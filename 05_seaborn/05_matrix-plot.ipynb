# 1. heatmap

A Heatmap is a graphical representation of data where values are represented using colors.

Why use Heatmaps?

Imagine a company has sales data.
```
Jan   1200
Feb   4500
Mar   900
Apr   8000
May   7000
```
Reading numbers:
```
1200
4500
900
8000
7000
```
takes time.

Heatmap:
```
Jan   ░
Feb   ▓
Mar   ░
Apr   █
May   █
```

Immediately you understand:

- April is highest
- March is lowest

That's the power of color.

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

sns.heatmap(data)

plt.show()
````

## 1.2 annot

- annot = Annotation
- Displays numerical values inside each heatmap cell.
- Default = False
- annot=True displays the values.

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

sns.heatmap(data, annot=True)

plt.show()
````

## 1.2 fmt


• Controls how annotation values are displayed.

Common formats:

fmt=".2f" → 2 decimal places

fmt=".1f" → 1 decimal place

fmt="d"   → Integer

````python
data = [
[0.8546321],
[0.1234567],
[0.9987543]
]

sns.heatmap(data, annot= True, fmt=".3f")
````

## 1.3 cmap

- cmap = Color Map
- Controls the colors of the heatmap.

```
cmap	Colors
"Blues"	    White → Blue
"Greens"	White → Green
"Reds"	     White → Red
"Purples"	White → Purple
"Oranges"	White → Orange
"coolwarm"	Blue ↔ White ↔ Red
"viridis"	Purple → Green → Yellow
"rocket"	Dark → Pink → White
```

1. Business Dashboard
- cmap="Blues"
2. Correlation Heatmap 
- cmap="coolwarm"
3. Scientific Visualization
- cmap="viridis"

````python
data = [
[0.8546321],
[0.1234567],
[0.9987543]
]

sns.heatmap(data, annot= True, fmt=".3f", cmap = 'coolwarm')
````

````python
data = [
[0.8546321],
[0.1234567],
[0.9987543]
]

sns.heatmap(data, annot= True, fmt=".3f", cmap = 'Blues')
````

## 1.4 linewidth


- Adds borders between cells.
- Default = 0
- Larger value = thicker borders.

````python
data = [
[0.8546321],
[0.1234567],
[0.9987543]
]

sns.heatmap(data, annot= True, fmt=".3f", cmap = 'Blues', linewidths=10, linecolor='black')
````

## 1.5 linecolor

- Controls border color.
- Used together with linewidths.

````python
sns.heatmap(
    data,
    linewidths=1,
    linecolor="black"
)
````

## 1.6 square=True

- Makes every heatmap cell square.
- Default = False.

````python
sns.heatmap(
    data, square=True
)
````

## 1.7 cbar

- Shows or hides the color bar.
- Default = True.

````python
sns.heatmap(
    data, cbar=False
)
````

# 2. Correlation Heatmap

• Correlation measures the strength and direction of the relationship between two variables.

Correlation Range:

- +1 → Perfect Positive Correlation
- 0 → No Correlation
- -1 → Perfect Negative Correlation

## 2.1 correlation matrix

- A correlation matrix shows the correlation between every pair of numerical variables in a dataset.


Properties:
- • Diagonal values are always 1.
- • The matrix is symmetric.

Example

Suppose we have 4 columns:

- Age
- Salary
- Experience
- Height

```

	Age	Salary	Experience	Height
Age	1.00	0.85	0.90	0.12
Salary	0.85	1.00	0.95	0.08
Exper	0.90	0.95	1.00	0.10
Height	0.12	0.08	0.10	1.00

```

````python
import pandas as pd

df = pd.DataFrame({
    "Age": [20,22,25,30],
    "Salary": [25000,30000,40000,50000],
    "Experience": [1,2,4,6]
})
df
````

````python
df.corr()
````

````python
import seaborn as sns
````

````python
corr = df.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap = 'coolwarm'
)
````

Creating a Correlation Heatmap

1. Calculate correlation matrix
   - df.corr()

2. Visualize using heatmap
```
   sns.heatmap(
       df.corr(),
       annot=True,
       cmap="coolwarm"
   )
```
Purpose:
- Quickly understand relationships between numerical features.
- Detect strong positive and negative correlations.
- Useful in EDA, Feature Engineering, and Machine Learning.

# 3. Clustermap

- Combines a heatmap with clustering.
- Groups similar rows and columns together.
- Displays dendrograms.
- Used mainly in research, biology, customer segmentation, and clustering analysis.

````python
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")

sns.clustermap(
    iris.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.show()
````

# 1. heatmap

A Heatmap is a graphical representation of data where values are represented using colors.

Why use Heatmaps?

Imagine a company has sales data.
```
Jan   1200
Feb   4500
Mar   900
Apr   8000
May   7000
```
Reading numbers:
```
1200
4500
900
8000
7000
```
takes time.

Heatmap:
```
Jan   ░
Feb   ▓
Mar   ░
Apr   █
May   █
```

Immediately you understand:

- April is highest
- March is lowest

That's the power of color.

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

sns.heatmap(data)

plt.show()
````

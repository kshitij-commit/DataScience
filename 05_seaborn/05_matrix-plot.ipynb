# 1. heatmap

A Heatmap is a graphical representation of data where values are represented using colors.

Why use Heatmaps?

Imagine a company has sales data.
```
Jan   1200
Feb   4500
Mar   900
Apr   8000
May   7000
```
Reading numbers:
```
1200
4500
900
8000
7000
```
takes time.

Heatmap:
```
Jan   ░
Feb   ▓
Mar   ░
Apr   █
May   █
```

Immediately you understand:

- April is highest
- March is lowest

That's the power of color.

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

sns.heatmap(data)

plt.show()
````

## 1.2 annot

- annot = Annotation
- Displays numerical values inside each heatmap cell.
- Default = False
- annot=True displays the values.

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

sns.heatmap(data, annot=True)

plt.show()
````

## 1.2 fmt


• Controls how annotation values are displayed.

Common formats:

fmt=".2f" → 2 decimal places

fmt=".1f" → 1 decimal place

fmt="d"   → Integer

````python
data = [
[0.8546321],
[0.1234567],
[0.9987543]
]

sns.heatmap(data, annot= True, fmt=".3f")
````

## 1.3 cmap

- cmap = Color Map
- Controls the colors of the heatmap.

```
cmap	Colors
"Blues"	    White → Blue
"Greens"	White → Green
"Reds"	     White → Red
"Purples"	White → Purple
"Oranges"	White → Orange
"coolwarm"	Blue ↔ White ↔ Red
"viridis"	Purple → Green → Yellow
"rocket"	Dark → Pink → White
```

1. Business Dashboard
- cmap="Blues"
2. Correlation Heatmap 
- cmap="coolwarm"
3. Scientific Visualization
- cmap="viridis"

````python
data = [
[0.8546321],
[0.1234567],
[0.9987543]
]

sns.heatmap(data, annot= True, fmt=".3f", cmap = 'coolwarm')
````

````python
data = [
[0.8546321],
[0.1234567],
[0.9987543]
]

sns.heatmap(data, annot= True, fmt=".3f", cmap = 'Blues')
````

## 1.4 linewidth


- Adds borders between cells.
- Default = 0
- Larger value = thicker borders.

````python
data = [
[0.8546321],
[0.1234567],
[0.9987543]
]

sns.heatmap(data, annot= True, fmt=".3f", cmap = 'Blues', linewidths=10)
````

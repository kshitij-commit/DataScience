# 1. histplot

- Used to visualize the distribution of numerical data.
- Groups data into intervals called bins.
- The height of each bar represents the frequency (count) of observations in that bin.

 Used for numerical data.
- Shows the distribution of data.
- Groups values into bins.
- Bar height = Frequency (Count).
- Bars touch each other because the data is continuous.

It helps answer questions like:

- Where are most values concentrated?
- Is the data normally distributed?
- Is the data left-skewed or right-skewed?
- Are there multiple peaks?
- Are there possible outliers?

```
sns.histplot(
    data=tips,
    x='total_bill'
)
```

````python
import seaborn as sns
````

````python
tips = sns.load_dataset(
    'tips'

)
````

````python
sns.histplot(
    data = tips,
    x = 'total_bill'
)
````

## 1.1 bins

```
5-10
10-15
15-20
20-25
25-30
```
These ranges are called bins.

One bar = One bin (range).

bins=10

means:

Divide the data into 10 intervals (10 bars).

- Bins are intervals (ranges) used to group numerical data.
- One bar represents one bin.
- More bins → More bars → More detail.
- Fewer bins → Wider bars → Less detail.
- Changing bins changes only the visualization, not the data.

````python
sns.histplot(
    data = tips,
    x = 'total_bill',
    bins = 5
)
````

````python
sns.histplot(
    data = tips,
    x = 'total_bill',
    bins = 20

)
````

## 1.2 Kde

What is KDE?

KDE = Kernel Density Estimation

KDE is a smooth estimate of the data distribution.

A KDE curve makes it easier to see:

- Where most values are concentrated.
- Whether the data is symmetric.
- Whether it's left-skewed or right-skewed.
- Whether there are multiple peaks.

 KDE stands for Kernel Density Estimation.

- It draws a smooth curve over the histogram.
- It helps understand the shape of the distribution.
- kde=True adds the density curve.
- It does not change the data or the histogram.

````python
"Example 1"

sns.histplot(
    data = tips,
    x = 'total_bill',
    kde = True
)
````

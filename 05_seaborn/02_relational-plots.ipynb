# Topic 1: sns.scatterplot()

Why do we use a Scatter Plot?

Scatter Plot is used to visualize the relationship between two numerical variables.

```
Suppose you have the tips dataset.

total_bill    tip
20            2
30            5
40            7
15            1

Using a scatter plot, you can answer:

Does a higher bill generally lead to a higher tip?
Are there any outliers?
Are there clusters?
Is there a positive or negative relationship?
```


## syntax

```
sns.scatterplot(data=df, x='column1', y='column2')
```


````python
import seaborn as sns
tips = sns.load_dataset('tips')
tips.head(4)
````

````python
import matplotlib.pyplot as plt
sns.scatterplot(
    data=tips,
    x = 'total_bill',
    y = 'tip'
)
plt.show()
````

## Topic: hue

hue → Used to divide the data into different categories using different colors.

-  Separates data into categories.
-  Assigns different colors automatically.
-  Creates legend automatically.
-  Mostly used with categorical columns.

1. Without hue:-

````python
import matplotlib.pyplot as plt
sns.scatterplot(
    data=tips,
    x = 'total_bill',
    y = 'tip'
)
plt.show()
````

2. with hue

````python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='sex'
)
````

````python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker'
)
````

````python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='day'
)
````

## Topic : style

style → Differentiates categories using different marker styles.

Unlike hue:

1. hue   → Different colors

2. style → Different marker shapes

 Separates categories using marker shapes.
- Mostly used with categorical columns.
- Can be combined with hue.

````python
tips.head(3)
````

````python
'''Example 1'''

sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    style = 'sex'

)
````

````python
'''Example 2'''

sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    style='sex',
    hue='sex'
)
````

## Topic : size

size → Represents a variable by changing the size of the markers.

 Changes marker size.
- Can represent numerical or categorical variables.
- Larger value → Larger marker.
- Can be combined with hue and style.

````python
tips.head(5)
````

1. Without size

````python
import seaborn as sns

sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip'
)
````

2. with size.

````python
""" Example 1"""

sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    size='tip'  # small tip small marker
)
````

````python
"""Example 2"""

sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    size = 'size'   # number of people. 
)
````

````python
"Example 3"

sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = 'tip',
    hue = 'sex',
    style = 'smoker',
    size = 'size' 
)
````

## Topic: palette

palette → Used to change the color theme for categorical variables.

Unlike color:

- color → Uses one color for all data.
- palette → Uses multiple colors for different categories.

palette

- Changes the color theme.
- Mostly used with hue.
- Different categories receive different colors.

Common palettes:
- deep
- Set2
- colorblind
- viridis

1. Without palette

````python
"Example 1"


import seaborn as sns
import matplotlib.pyplot as plt 


sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'sex'
)
````

````python
tips.head(3)
````

2. With palette

````python
"""Example 2"""


sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'sex',
    palette='Set2'
)
plt.show()
````

````python
"""Example 3"""




sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'sex',
    palette='colorblind'
)
plt.show()
````

````python
"""Example 4"""




sns.scatterplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'day',
    palette='Set2'
)
plt.show()
````

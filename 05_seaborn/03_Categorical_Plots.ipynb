# Topic 1: countplot()

countplot() → Displays the count (frequency) of each category.

 Used for categorical data.
- Counts the number of observations in each category.
- Creates bars automatically.
- No need to calculate counts manually.

```
Syntax
sns.countplot(
    data=df,
    x='column_name'
)
```

````python
import matplotlib.pyplot as plt 
import seaborn as sns

tips = sns.load_dataset('tips')
tips.head(3)
````

````python
sns.countplot(
    data = tips,
    x = 'sex'
              )

plt.show()
````

````python
sns.countplot(
    data=tips,
    x = 'day'
)
````

## hue

hue in countplot()

- Splits each category into subcategories.
- Uses different colors for each subcategory.
- Automatically creates a legend.

````python
"Example 1"
````

````python
sns.countplot(
    data = tips,
    x ='sex',
    hue = 'smoker'
)
````

## order

- Controls the order of categories.
- Accepts a list.
- Useful when default ordering is not suitable.

````python
tips.head(2)
````

````python
sns.countplot(
    data = tips,
    x = 'day',
    order=['Mon', 'Sun', 'Thur', 'Fri', 'Sat']

)
````

````python
sns.countplot(
    data = tips,
    x ='sex',
    order=['Male', 'Female']
)
````

# Topic 2: barplot()

barplot() → Used to compare the average (or another aggregated value) of a numerical variable across different categories.

Calculates an aggregated numerical value (default = mean)

- Used to compare average numerical values across categories.
- x → Categorical column
- y → Numerical column
- Default estimator = mean()
- Height of each bar = average value.

````python
sns.barplot(
    data = tips,
    x = 'sex',
    y = 'tip'
)
plt.show()
````

## estimator

````python
sns.barplot(
    data = tips,
    x = 'sex',
    y = 'tip',
    estimator='median'
)
plt.show()
````

## hue

Splits each category into subcategories.

- Calculates the average (default estimator) for each subcategory.
- Different colors represent different subcategories.
- Creates the legend automatically.

````python
sns.barplot(
    data = tips,
    x = 'day',
    y = 'tip',
    hue = 'sex'
)

plt.show()
````

## palette

palette → Changes the color theme of the bars.

palette

- Changes the color theme.
- Mostly used with hue.
- Common palettes:
  1. deep
  2. Set2
  3. colorblind
  4. viridis

````python
sns.barplot(
    data = tips,
    x = 'day',
    y = 'tip',
    hue = 'smoker',
    palette='Set1'
)
plt.legend(loc = 'upper right')
plt.show()
````

## order

- Controls the order of categories.
- Accepts a list.

````python
sns.barplot(
    data = tips,
    x = 'day',
    y = 'tip',
    order = ['Sun', 'Thur', 'Fri', 'Sat']
)
````

# Boxplot

boxplot() → Used to visualize the distribution of a numerical variable across different categories using the five-number summary.

 Compares the distribution of numerical data across categories.

• Shows:
  1. Minimum
  2. Q1
  3. Median
  4. Q3
  5. Maximum
  6. Outliers
- x → Categorical column
- y → Numerical column

```
Syntax
sns.boxplot(
    data=df,
    x='Category',
    y='Numerical Column'
)
```


````python
import matplotlib.pyplot as plt 
import seaborn as sns
````

````python
tips = sns.load_dataset('tips')
````

````python
tips.head(3)
````

````python
sns.boxplot(
    data = tips,
    x = 'day',
    y = 'total_bill'
)
plt.show()
````

## hue

- Splits each category into subcategories.
- Draws a separate boxplot for each subcategory.
- Helps compare distributions between subcategories.
- Creates the legend automatically.

````python
sns.boxplot(
    data = tips,
    x = 'day',
    y = 'total_bill',
    hue = 'sex'

)
plt.show()
````

## palette



- Changes the color theme.
- Mostly used with hue.
- Does not change the data.

````python
sns.boxplot(
    data=tips,
    x='day',
    y='total_bill',
    palette='Set2'
)
````

````python
sns.boxplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='sex',
    palette='colorblind'
)
````

# Violinplot

- Used to compare the distribution of numerical data across categories.
- Shows everything a boxplot shows.
- Also displays the density (shape) of the distribution.

- Shows distribution across categories.
- Includes boxplot information.
- Also shows density (shape) of the distribution.
- Wider section → More observations.
- Narrower section → Fewer observations.

The wider the violin,

➡️ the more data points are concentrated there.

The narrower the violin,

➡️ the fewer data points are there.

````python
sns.violinplot(
    data = tips,
    x = 'sex',
    y = 'tip'
)
````

````python
sns.boxplot(
    data=tips,
    x='sex',
    y='tip'
)
````

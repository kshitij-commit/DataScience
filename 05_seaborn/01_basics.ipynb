# Topic 1: What is Seaborn?

Seaborn is a Python data visualization library built on top of Matplotlib.

It provides:

- Better-looking graphs by default.
- Simpler syntax.
- Better support for Pandas DataFrames.
- Built-in statistical visualizations.

## Why Seaborn?

Suppose you want to create a box plot.

1. Matplotlib
- plt.boxplot(data)
2. Seaborn
- sns.boxplot(data=data)

Q: Can Seaborn work without Matplotlib?

Answer:

No. Seaborn is built on top of Matplotlib and uses it internally to render plots.

# Topic 4: Exploring a Dataset

```
1. head()

Shows the first 5 rows.

tips.head()
Notes
head() → Displays the first 5 rows of the dataset.
2. info()

Shows:

Number of rows
Number of columns
Data type of each column
Missing values
Memory usage
tips.info()
Notes
info() → Displays dataset information.

• Total entries
• Data types
• Non-null values
• Memory usage
3. describe()

Shows statistics for numerical columns.

tips.describe()

Output includes:

count
mean
std
min
25%
50%
75%
max
Notes
describe() → Statistical summary of numerical columns.
4. columns

Displays all column names.

tips.columns
Notes
columns → Displays all column names.
5. dtypes

Shows the data type of every column.

tips.dtypes

Example:

total_bill    float64
tip           float64
sex           category
day           category
Notes
dtypes → Displays the data type of each column.
```

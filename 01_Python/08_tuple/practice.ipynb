# Tuple
A Tuple is an ordered collection of elements that is immutable, meaning its elements cannot be changed after creation.

Characteristics:

1. Ordered
2. Unchangeable
3. Allow duplicates

## Basic of Tuples

### Creating tuples

```python
"Example 1"

fruits = ("apple", "banana", "grapes")

print(fruits)
```

```python
"Example 2"

a = (1,"apple")
a
```

#### Tuple as single element

1. Tuple != (  )  parenthesis

```python
t = (1)

print(type(t))
```

Because Python sees:

(1)

as just:

1

The parentheses are only grouping symbols here.

```python
" Example 1"

l = (1+2)

print(type(l))

""" Python doesn't think this is a tuple. """
```

```python
"Example 2"


fruit = ("apple")

print(type(fruit))

""" This is not a tuple ha.. ha.. ha.. """
```



##### Tuple = Comma ,



```python
"Example 2"


num = 1,2,3

print(type(num))

"""
Even without parentheses!

Because the commas create the tuple.
"""
```

```python
"Example 3 "


t = (1,)

print( type (t))
```

### Operations on tuples

Just like strings and lists, tuples support some basic operations.

#### 1. Cancatenation (+)

```python
t1 = (1,2,3,4)
t2 = (4,5,6)
print(t1+t2)

""" concatenate not addition"""
```

#### 2. Repetition (*)

Used to repeat a tuple.

It repeats the tuple specified number of times.

```python
t1 = (1,2,3)

print(t1 *3)
```

#### 3. Membership operation (in)

```python
print(t)
```

```python
1 in t
```

```python
print(40 not in t)
```

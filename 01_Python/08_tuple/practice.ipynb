# Tuple
A Tuple is an ordered collection of elements that is immutable, meaning its elements cannot be changed after creation.

Characteristics:

1. Ordered
2. Unchangeable
3. Allow duplicates

### Operations on tuples

Just like strings and lists, tuples support some basic operations.

#### 1. Cancatenation (+)

```python
t1 = (1,2,3,4)
t2 = (4,5,6)
print(t1+t2)

""" concatenate not addition"""
```

#### 2. Repetition (*)

Used to repeat a tuple.

It repeats the tuple specified number of times.

```python
t1 = (1,2,3)

print(t1 *3)
```

#### 3. Membership operation (in)

```python
print(t)
```

```python
1 in t
```

```python
print(40 not in t)
```

## Tuple unpacking 

```python
""" tuples unpacking"""
 
a,b,c = (1,2,3)

print(a,b,c)
```

```python
a,b = (1,2,3)
```

```python
""" swap two values """

a = 1
b = 2
print(a,b)

a,b = b,a
print(a, b)
```

```python
a,b, *others = (1,2,3,4,5)


print(a,b)
print(others)
```

```python
## Zipping tuples

t1 = (1,2,3,4) 
t2 = (1,4,9,16)
zip(t1,t2)
```

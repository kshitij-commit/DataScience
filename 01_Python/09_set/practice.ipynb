# Set

A Set is an unordered, collection of items. Every set element is unique (no duplicates) must be immutable (cannot be changed). 

However, a set itself is  mutable, we can add or remove items from it. 

Characteristics:- 

1. Unordered
2. Mutable
3. No duplicates
4. Can't contain mutable datatypes. 

## 1. Creating sets 

````python
" Example 1"

s = {}   # curly braces means dictionary.

print(type(s))
````

### a. Correct impty set 

````python
a = set()

print(type(a))
````

### b. Homogeneos and hetrogenous sets


````python
" Example 1"

s1 = {1,2,3,4,5}

print(s1)
````

````python
"Example 2"

s2 = {1,2,(3,4,5), "Python"}  ## sets does't contain mutable data types.

print(s2)
print(type(s2))
````

````python
"Example 3"

s3 = {1, True}  # True = 1 , so {1,1} => {1}
print(s3)
````

### c. type convertion

````python
" Example 1 "

s4 = set([1,2,3,4])
print(s4)
````

````python
""" duplicates not allowed """

s5 = {1,2,2,3,3,3,4,5}

s5 
````

````python
""" Unordered """

s6 = {1,2,3}
s7 = {3,2,1}

print(s6 == s7)
````

````python
"Example 2"

s = set("Python")
print(s)
````

## 2. Accessing items


````python
" Example 1 "

""" Can we access set items using indexing?"""

s = {10,20,30}
print(s[0])

````

```
why ?
Because sets are unordered.
s = {10,20,30}
There is no guaranteed order.
```

### Can we use slicing?

````python
" Example 1"

s = {10,20,30}

print(s[:])
````

```
Same reason.

No indexing → No slicing.
```

### Then how do we access items?

#### Method 1: Using Loops

````python
"Example 1"

s = {1,2,3,4}

for i in s:
    print(i)
````

````python
#### Method 2: Membership Operator

s = {1,2,3,4}

print( 2 in s)
print( 30 in s)
````

## 3. Edit items

### add items

## 4. Deleting items

### 1. Remove()
Used to remove a specific element.

````python
" Example 1 "

s1 = {1,2,3,4}

print(s1)

s1.remove(2)

print(s1)
````

````python
s1.remove(5)
````

### 2. discard()

Also removes an element.

````python
"Example 1"

s1 = {1,2,3,4}

print(s1)

s1.discard(2)

print(s1)
````

````python
print(s1.discard(5))
````

### 3. del

Deletes entire set.

````python
" Example 1 "

s1 = {1,2,3,4}

print(s1)
del s1
print(s1)
````

### 4. pop()

Randomly deletes an item.

````python
" Example 1 "

s1 = {1,2,3,4,5}

print(s1)

s1.pop()
print(s1)

s1.pop()
s1.pop()

print(s1)



````

### 5. clear()

Deleted all elements.

````python
"Example 1"

s1 = {1,2,3,4,5}

print(s1)
s1.clear()
print(s1)
````

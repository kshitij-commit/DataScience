# Set

A Set is an unordered, collection of items. Every set element is unique (no duplicates) must be immutable (cannot be changed). 

However, a set itself is  mutable, we can add or remove items from it. 

Characteristics:- 

1. Unordered
2. Mutable
3. No duplicates
4. Can't contain mutable datatypes. 

## Unordered

No indexing. We can't access elements using indexing.

```python
" Example 1"

s = {1,2,3}

s[0]
```

## 1. Creating sets 

```python
" Example 1"

s = {}   # curly braces means dictionary.

print(type(s))
```

### a. Correct impty set 

```python
a = set()

print(type(a))
```

### b. Homogeneos and hetrogenous sets


```python
" Example 1"

s1 = {1,2,3,4,5}

print(s1)
```

```python
"Example 2"

s2 = {1,2,(3,4,5), "Python"}  ## sets does't contain mutable data types.

print(s2)
print(type(s2))
```

```python
"Example 3"

s3 = {1, True}  # True = 1 , so {1,1} => {1}
print(s3)
```

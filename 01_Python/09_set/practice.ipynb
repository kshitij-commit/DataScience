# Set

A Set is an unordered, collection of items. Every set element is unique (no duplicates) must be immutable (cannot be changed). 

However, a set itself is  mutable, we can add or remove items from it. 

Characteristics:- 

1. Unordered
2. Mutable
3. No duplicates
4. Can't contain mutable datatypes.

## 5. Sets Operations

### 1.  Union (|)

Combine all unique elements from both sets.

````python
"Example 1"

s1 = {1,2,3,4,5,5,2,3,4}

s2 = {1,2,50,31,3,4,5,5,33}


print(s1 | s2)
````

### 2. Intersection (&)

Common elements in both sets.

````python
"Example 1"

s1 = {1,2,2,3,4,50, 40, 5}
s2 = {5,5,4,3,2,1}

print(s1 & s2)
````

### 3. Difference (-)

Elements present in first set but NOT in second set.

````python
s1 = {1,2,3,4}
s2 = {1,2,5,4}

print(s1 - s2)
````

### 4.Symmetric Difference (^)

Elements that are in either set, but NOT in both.

Common elements remove.

````python
"Example 1"

A = {1, 2, 3}
B = {3, 4, 5}

print(A ^ B)
````

### 8. union()

````python
"Example 1"

s1 = {1,2,3,4}
s2 = {3,4,5,6,7}

print("Before union:-")
print("s1:",s1)
print("s2:",s2)
s1 = s1.union(s2)
print("After union:-")
print("s1:",s1)
print("s2:",s2)


# s1.update(s2)
````

### 10. intersection/intersection_update

````python
"Example 1"

s1 = {1,2,3,4}
s2 = {3,4,5,6}

print("Before intersection:-")
print("s1:",s1)
print("s2:",s2)

s1 = s1.intersection(s2)
print("After intersection:-")
print("s1:",s1)
print("s2:",s2)
````

````python
"Example 2"

s1 = {1,2,3,4}
s2 = {3,4,5,6}

print("Before intersection:-")
print("s1:",s1)
print("s2:",s2)

s1.intersection_update(s2)
print("After intersection:-")
print("s1:",s1)
print("s2:",s2)


````

### 11. isdisjoint()

Checks whether two sets have no common elements.

````python
"Example 1"

A = {1, 2}
B = {3, 4}

print(A.isdisjoint(B))
````

````python
"Example 2"

A = {1, 2}
B = {2, 3}

print(A.isdisjoint(B))
````

### 12. issubset()

Checks whether all elements of one set are present in another set.

````python
"Example 1"

A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))
````

````python
"Example 2"

A = {1, 5}
B = {1, 2, 3, 4}

print(A.issubset(B))
````

### 13. issuperset()

Opposite of subset.

Checks whether a set contains all elements of another set.

````python
"Example 1"

A = {1, 2, 3, 4}
B = {1, 2}

print(A.issuperset(B))
````

````python
"Example 2"

A = {1, 2}
B = {1, 2, 3}

print(A.issuperset(B))
````

````python
print(B.issuperset(A))
````

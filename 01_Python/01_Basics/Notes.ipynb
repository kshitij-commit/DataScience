Data Science is the process of extracting useful insights and predictions from data using statistics, programming, and machine learning

### 1. Algorithm
Step-by-step procedure to solve a problem.

Example
1. Input two numbers
2. Add them
3. Print result
 
 Algorithm is a finite step-by-step procedure used to solve a problem.

### 2. Flowchart
Graphical representation of an algorithm.

Symbol	Meaning
1. Oval  ->	Start/Stop
2. Rectangle ->	Process
3. Parallelogram ->	Input/Output
4. Diamond ->	Decision

Flowchart helps visualize program logic.

### 3.Pseudocode
English-like representation of program logic.

Example:-
1. START
2. INPUT A, B
3. SUM = A + B
4. PRINT SUM
5. END

Important Points
1. easy to understand
2. language-independent
3. helps before coding

Pseudocode is an informal way of writing program logic.


### 4.Programming
Writing instructions for computers using programming languages.

Why Programming is Used
1. automation
2. problem solving
3. application development
4. data processing

Programming is used to solve problems and automate tasks.

### 5. Difference Between Algorithm and Program
```

Algorithm	                  Program
Logic	                      Implementation
Step-by-step solution	      Executable code
Language-independent	      Written in programming language
```

# What makes python special?

### 1. Module
A module is a single Python file containing code, functions, or variables.

A module is a single Python file containing reusable code.
```
Example:

math.py

Inside it:

def add(a,b):
    return a+b

That single file = module.
```

### 2. Package
A package is a collection of related modules.
```
Example:
mypackage/
    math.py
    stats.py
    graphs.py

Together = package.
Package = folder containing modules.
```

### 3. Library
A library is a collection of packages and modules used for specific tasks.

Example:
1. NumPy
2. Pandas
3. Matplotlib

### 4. PyPI
Python Package Index

PyPI is the official repository/storage location for Python packages.

Simple meaning:

It is a place from where Python libraries/packages can be downloaded.

Example

When you write:

pip install numpy

pip downloads NumPy from PyPI.

#####   Important Terms
```
Term        	Meaning
pip	            package installer
PyPI	        online repository of Python packages
```

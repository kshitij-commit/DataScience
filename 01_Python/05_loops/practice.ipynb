# Loop

A loop is used to execute a block of code repeatedly until a condition is satisfied. 

## while loop

A while loop executes a block of code repeatedly as long as the condition remains True.

```
Syntax

while condition:
    statement
```

````python
"""Example 1"""


n = 1

while n<=5:
    print(n)
    n+=1
````

```

Iteration

One complete execution of the loop body.
```

````python
"""Example 2"""


n = 1

while n<=3:
    print("Hello")
    n+=1
````

```
What is an Infinite Loop?

An infinite loop is a loop that never terminates because its condition always remains True.
```

### Loop Control Variable

```
i = 1

while i <= 5:
    print(i)
    i += 1

i is called the loop control variable.

It controls how long the loop runs.
```

````python
"""Example 3"""

"""Take a value and write its multiplocation till 10. """


num = int(input("Enter a number: "))

i = 1
while i <= 10:
    print(f"{num} * {i} = {num*i}")
    i+=1
````

## Break Statement

The Break statement is used to immediately terminate a loop.  

```
Syntax
while condition:
    statement

    if condition:
        break
```

````python
""" Example 1"""


i = 1
while i < 10:
    print(i)

    if i == 5:
        break
    i+=1
````

```
Does break terminate only the current iteration?

Answer:

No. break terminates the entire loop.
```

## Continue statement
The continue statement skips the current iteration and moves directly to the next iteration of the loop.

### Difference Between Break and Continue
```
break
break

👉 Exit the entire loop.

continue
continue

👉 Skip the current iteration and continue with the next one.
```

````python
"""Example 1"""


i = 1
while i <= 10:
    if i == 5:
        i+=1     # write here because without this i never becomes 6 and loop continusly runs on 5.
        continue

    print(i)

    i = i + 1


````

## For loop

A for loop is used to iterate over a sequence and execute a block of code for each item.

```
Syntax
for variable in sequence:
    statement
```

```
What Can a For Loop Iterate Over?
String
List
Tuple
Set
Dictionary
Range

These are called iterables.
```

````python
"""Example 1"""

""" Write a program to print and add square of the number to a list  till 10.  """
````

````python
l = []

for i in range(10):
    print(i+1)
    l.append(pow(i,2))
print(l) 
````

````python
""" Example 2 """

""" Write a program to separate negative and positive integers from the list.  """

l = [24, -15, 89, 12,36,-85, -2,-9,-31,32,14]
l1 = []
l2 = []
for i in l:
    if (i < 0):
        l1.append(i)
    else:
        l2.append(i)
print(l1)
print(l2)
````

````python
""" Example 2 """

""" Write a program to take a string from user and count vowels in it. """

name = input("Enter a string: ")

count = 0
for i in name:
    if (i == 'a' or i == 'i' or i == 'o' or i == 'u' or i == 'e'):
        count+=1
        # print(i)

print(f"String >> {name} << has {count} vowels.")
````

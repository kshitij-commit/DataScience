## Bitwise Operator
In Python bitwise operators are used to perform bitwise calculations on integers. The integers are first converted into binary and then operations are performed on each bit or corresponding pair of bits.The result is then returned in decimal format.

```
Operator                Meaning
&                       Bitwise AND
|                       Bitwise OR
^                       Bitwise exclusive OR/Bitwise XOR
~                       Bitwise inversion(one's complement)
<<                      Shift the bit to left/Bitwise Left shift
>>                      Shift the bits to right/Bitwise Right Shiftm



## Bitwise AND 

````python
a=5
b=4

print(a&b)

````

```
a = 5
b = 4
1. a in binary => 0 1 0 1
2. b in binary => 0 1 0 0

& bitwise and  gives (1 1)=> 1 else 0

           0 1 0 1
           0 1 0 0

result     0 1 0 0 => 4 
```

````python
x=125
y=1593
print(x & y)
print(type(x & y))
````

## Bitwise Inversion

````python
res=~5
print(res)
````

1. Operation: Flips every 0 to 1 and every 1 to 0.
2. Operator: represented by ~ (tilde) in most languages.

#### Key Formulas (How to Calculate)

1. For Signed Integers  Computers use Two's Complement for signed numbers. 

The shortcut formula is: 
##### ~x=-x-1
Example
1. ~5 = -5 - 1 = -6
2. ~-10 = -(-10) - 1 = 9 

##### 2's complement = 1's complement + 1

````python
a = 10
print(~a)
````

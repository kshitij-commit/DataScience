# Dictionary

A Dictionary is a mutable collection of key-value pairs.

Characteristics:-
1. Mutable
2. No Indexing
3. Keys can't be duplicated
4. Keys can't be mutable items.  

A dictionary stores data as:

key : value

Dictionaries do not support numeric/positional indexing. Data is accessed using keys.

Used to delete a specific key-value pair.

````python
student = {
    "name": "Ram",
    "age": 20
}

del student["age"]

print(student)
````

````python
"""
What if Key Doesn't Exist?
del student["city"]

Output:

KeyError

Because "city" is not present.
"""
````

### 2. pop()

Removes a key and returns its value.

````python
"Example 1"

student = {
    "name": "Ram",
    "age": 20
}

x = student.pop("age")

print(x)
print(student)
````

````python
"""
What if Key Doesn't Exist?
student.pop("city")

Output:

KeyError

"""
````

## 9. Dictionary Comprehension 
Dictionary Comprehension is a concise way to create a dictionary in a single line.

```
Syntax

{key: value for item in iterable}

```

````python
"Example 1"

d = {x: x*x for x in range(5)}

print(d)
````

````python
"Example 2"

d = {x: x*x for x in range(10) if x % 2 == 0}

print(d)
````

#### zip()

````python
"Example 1"

month = ["jan", "feb", "march","april"]
days = [31,28,31,30]

{i : j for  (i,j) in zip(month,days) }
````

````python
"Example 2"

d2 = {1:0,2:0,3:9,4:0, 5:12}

d3 = {x : y for (x,y) in d2.items() if y>0}
print(d3)
````

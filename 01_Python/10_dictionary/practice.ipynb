# Dictionary

A Dictionary is a mutable collection of key-value pairs.

Characteristics:-
1. Mutable
2. No Indexing
3. Keys can't be duplicated
4. Keys can't be mutable items.  

A dictionary stores data as:

key : value

Dictionaries do not support numeric/positional indexing. Data is accessed using keys.

## 1. Properties of Dictionary

### 1. Mutable

```python
"Example 1"

dict1 = {"Name":"Nitish", "Age":33, "gender":"Male"}
dict1
```

```python
dict1["Name"]  = "rossom"
dict1
```

### 2.Keys Must Be Unique

```python
"Example 1"

d = {
    "name": "A",
    "name": "B" 
}



print(d)
```

```python
""" the last value wins"""
```

### 3. Why Dictionary Keys Must Be Immutable?

Dictionary keys must be immutable (hashable) because Python uses keys for fast lookup using hashing.

### 3.  2d dictionary

A 2D Dictionary is a dictionary whose values are themselves dictionaries.

```python
"Example 1"

students = {
    101: {
        "name": "Ram",
        "age": 20
    },

    102: {
        "name": "Shyam",
        "age": 21
    }
}
print(students)
```

```python
print(students[101]["name"])
```

## 5. Remove key-value pair.

### 1. del statement
Used to delete a specific key-value pair.

```python
student = {
    "name": "Ram",
    "age": 20
}

del student["age"]

print(student)
```

```python
"""
What if Key Doesn't Exist?
del student["city"]

Output:

KeyError

Because "city" is not present.
"""
```

### 2. pop()

Removes a key and returns its value.

```python
"Example 1"

student = {
    "name": "Ram",
    "age": 20
}

x = student.pop("age")

print(x)
print(student)
```

```python
"""
What if Key Doesn't Exist?
student.pop("city")

Output:

KeyError

"""
```

### 3. popitem()

popitem() removes and returns the last inserted key-value pair as a tuple.

```python
"Example 1"

student = {
    "name": "Ram",
    "age": 20
}

student.popitem()
```

```python
print(student)
```

### 4. clear()

Removes all key-value pairs.

```python
"Example 1"

student = {
    "name": "Ram",
    "age": 20
}

student.clear()

print(student)
```

### 5. del 
 
deletes entire Dictionary


```python
student = {
    
    "name": "Ram"
}

del student
```

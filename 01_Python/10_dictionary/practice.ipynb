# Dictionary

A Dictionary is a mutable collection of key-value pairs.

Characteristics:-
1. Mutable
2. No Indexing
3. Keys can't be duplicated
4. Keys can't be mutable items.  

A dictionary stores data as:

key : value

Dictionaries do not support numeric/positional indexing. Data is accessed using keys.

## 1. Properties of Dictionary

### 1. Mutable

```python
"Example 1"

dict1 = {"Name":"Nitish", "Age":33, "gender":"Male"}
dict1
```

```python
dict1["Name"]  = "rossom"
dict1
```

### 2.Keys Must Be Unique

```python
"Example 1"

d = {
    "name": "A",
    "name": "B" 
}



print(d)
```

```python
""" the last value wins"""
```

## 2. Creating dictionary

### 1. Empty Dictionary

### 2. using dict()

```python
d = dict()


print(d)
```

### 3.  2d dictionary

A 2D Dictionary is a dictionary whose values are themselves dictionaries.

```python
"Example 1"

students = {
    101: {
        "name": "Ram",
        "age": 20
    },

    102: {
        "name": "Shyam",
        "age": 21
    }
}
print(students)
```

```python
print(students[101]["name"])
```

### 4. Using Sequence and dict() Function


What does dict() expect?

Answer:

It expects key-value pairs.

```python
d = dict([
    ("name", "Ram"),
    ("age", 20)
])

print(d)
```

```python
"Example 2"

d = dict([
    ("name", "Ram"),
    ("age", 20)
])

print(d)
```

### 5. Mutable Items as Keys

Dictionary Keys Must Be Immutable.

Why?

Lists are mutable.

Dictionary keys must be immutable (hashable).

```python
"Example 1"

d = {
    1: "One",
    "name": "Ram",
    (1, 2): "Tuple"
}
d
```

```python
"Example 2"

d = {
    [1,2]:(1,4)  # keys can't be mutable items.

}
```

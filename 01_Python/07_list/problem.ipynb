# Question 1
Take a input of list from the user and then reverse the list.

```python
num = []
length = int(input("Enter length of list: "))
for i in range(length):
    num.append(input("Enter list elements: "))

print("Original list: ",num)
num.reverse()
print("Rversed list: ",num)
```

# Question 2

 Take a list as  input and interchanage its first and last elements.

```python
l = []
length = int(input("Enter length of list: "))
for i in range(length):
    l.append(input("Enter list elements: "))
print("Original list:", l)

temp = l[0]
l[0] = l[length-1]
l[length-1] = temp
print("Interchanges list: ",l)
```

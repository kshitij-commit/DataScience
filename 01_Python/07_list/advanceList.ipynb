# List comprehension

List Comprehension is a concise way to create a new list using a single line of code.

```
Basic Syntax

[expression for item in iterable]

````python
"Example 1"

num = [i for i in range(5)]
print(num)
````

````python
"""Example 2"""

squares = [i*i for i in range(5)]

print(squares)
````

List Comprehension With Condition


````python
"Example 1"

even = [i for i in range(10) if i % 2 == 0]
even
````

````python
"Example 2"

string = "python"

l = [ch for ch in string]
print(l)
````

````python
"Example 3"

names = ["python", "cpp", "javascript"]

l1 = [name.upper() for name in names]
print(l1)
````

## Advantage of list comprehension

1. More time-efficient and space-efficient.
2. Require fewer line of code.
3. Transform iterative statement into formula.

````python
"Example 4"

"Scalar multiplication on vector."

v = [1,2,3]

res = [i*2 for i in v  ]
print(res)
````

````python
"Example 5" 

""" Find languages which starts with p. """

lang = ["java", "Python", "Php", "c", "javascript"]

res = [name.lower() for name in lang if name.lower().startswith("p") ]

print(res)
````

````python
"Example 6"


basket = ["basket", "aguava", "cerry", "banana", "apple"]
my_fruits = ["apple", "kiwi", "grapes", "banana"]

""" add new list from my_fruits and items if the fruit exist in basket and also starts with "a". """

newl = [fruit for fruit in my_fruits if (fruit in basket) and fruit.startswith("a")    ]

print(newl)
````

````python
""" Example 7"""

""" print a (3,3) matrix using list comprehension -> nested list comprehension"""

l = [[i for i in range(3)] for j in range(3) ]
print(l)
````

````python
" Example 8"

"cartesian product of two list"

l1 = [1,2,3]
l2 = [4,5,6]

prod = [i*j for i in l1 for j in l2 ]
print(prod)
````

# Zip

zip() is a built-in function that combines elements from multiple iterables (lists, tuples, strings, etc.) position by position.

1. zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together.
2. If the passed iterator have different lengths, the iterator with the least items decides  the length of the new iterator.  

```
Basic Syntax

zip(iterable1, iterable2)

````python
"Example 1"

names = ["Ram", "Shyam", "Mohan"]
marks = [85, 90, 78]

res = zip(names, marks)


print(list(res))
res
````

```
important :-

zip() -> itself returns  <class 'zip'>

````python
print(tuple(res))
print(dict(res))
````

```
if already unzipped then zip return only impty list, tuple, dictionary.  

````python
"Example 2"


names = ["Ram", "Shyam", "Mohan"]
marks = [85, 90, 78]

res = (zip(names, marks))

print(dict(res))


````

````python
"Example 3"

names = ["Ram", "Shyam", "Mohan"]
marks = [85, 90, 78]

for i,j in zip(names,marks):
    print(i,j)
````

````python
"Example 4"

names = [1,2,3]
marks = [85, 90, 78]

res = [i+j for i,j in zip(names,marks)]
print(res)
````

````python
"Example 5"

a = [1,2,3]
b = ["a",'b']

res = zip(a,b)

print(tuple(res))
````

````python
"Example 6"

names = ["Ram","Shyam","Mohan"]
marks = [80,90,70]
cities = ["Delhi","Mumbai","Pune"]

print(list(zip(names, marks, cities)))
````

## Unzipping

The * is called unpacking.

````python
data = [('Ram',80), ('Shyam',90)]

names, marks = zip(*data)

print(names)
print(marks)
````

# Disadvantage of list

1. Lists Are Mutable

````python
"Example 1"

# Lists can be changed after creation.

nums = [1,2,3]

nums[0] = 100

print(nums)
````

Why is this a disadvantage?

Sometimes data should not change.

Example:

1. Days of week
2. Months
3. Mathematical constants

For such cases, Tuple is safer.

2. More Memory Consumption

Lists store extra information to allow modifications.

1. ist > Tuple

in memory usage.

Lists consume more memory than tuples because they are mutable.

3. Slower Than Tuple

Because lists can grow, shrink, insert, delete, etc.

Python needs extra work.

4. Risk of Accidental Modification

Example:

students = ["Ram", "Shyam"]

students[0] = "Mohan"

Now original data is changed.

In large projects this can create bugs.

5. Searching Is Slow in Large Lists

Example:

1000000 elements

To find an item, Python may need to check many elements.

"Is mutability always a disadvantage?"

Answer:

No.

It's also the biggest advantage of lists.

Because we can:

1. append()
2. insert()
3. remove()
4. pop()

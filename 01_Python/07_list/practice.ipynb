# What is a Data Structure?

A Data Structure is a way of organizing and storing data so that it can be accessed and modified efficiently.

```
Intuition

Suppose you have 500 student names.

Storing them like this:

student1 = "Rahul"
student2 = "Aman"
student3 = "Riya"

is not practical.

We need a way to store multiple values together.

That is why data structures exist.
```

# List

List is an ordered collection of multiple items stored in a single variable.

1. A collection means multiple values stored together.
2. Ordered means every element has a fixed position.
3. It is mutable. -> we can modify its elements after it has been formed. 
4. Allow duplicates.

## slicing on list

Slicing is used to access a portion (sublist) of a list.

Instead of getting a single element, slicing returns multiple elements.

### Syntax

```
list_name[start : stop:jump]

1. start → Included
2. stop → Excluded

````python
" Exampple 1 "


num = [10,20,30,40]
num[1:3]
````

````python
" Example 2"


nums = [10, 20, 30, 40, 50]

print(nums[-3:])
````

````python
" Example 3 "

nums = [10, 20, 30, 40, 50, 60]

print(nums[::2])
````

````python
" Example 4"

nums = [10, 20, 30, 40, 50,60]

print(nums[::-1])
````

## list methods


### 1) append()
The append() method is used to add a single element at the end of a list.

#### Syntax
```
list_name.append(element)

````python
" Example 1 "

nums = [10, 20]

result = nums.append(30)

print(result)
````

```
**************************************************************
 why None?


append() modifies the original list and returns None.

````python
" Example 1 "

nums = [10, 20]

nums.append(30)

nums
````

````python
" Example 2 "

a = [1,2]

a.append([3,4])
a
````

````python
" Example 3 "

b = [5,6]
print(b.append(3))
````

### 2) extend()
The extend() method is used to add multiple elements from an iterable to the end of a list.

### Syntax
```
list_name.extend(iterable)

````python
" Example 1 "


nums = [1,2]

nums.extend([4,5,6])
nums
````

````python
" Example 2 "

num = [3,4]

""" extend() works with any iterable."""

""" 1) List"""

num.extend([5,6])

print("iterable was list:",num)

""" 2) tuple """
num.extend((7,8))
print("itrable was tuple: ",num)

""" 3) string """
 
num.extend("ABCD")
print("iterable was string: ",num)
````

```
***************************************************************
                            Question 1.

````python
" Example 3 "

fruit = ["apple"]

fruit.extend("banana")
fruit
````

````python
" Example 4 "

text1 = ["data"]
text2 = ["science"]

text1.extend(text2)
print(text1)
print(text2)
````

```python
"""Question 1"""


""" A toy vendor supplies 3 types of toys. Battery based toys , key based toys, and Electrical charging based toys. The vendor gives a discount of 10% 
    on orders of battery based toys if the order is for more than Rs. 1000. On order for more than Rs. 100 for key based toys , a discount of 5% is given, 
    and a discount of 10%  is given on orders for electrical charging based toys of value more than Rs. 500. 
    Assumes that the numeric code 1,2,and 3 are used for battery based toys, key based tloys and electrical charging based toys respectively. 
    Write a program that reads 
    the product codes and the order amount and prints out the net amount that the customer is required to pay after the discount.
"""
```

```python
def battery(order, order_amt,net_amt):
    if order_amt>1000:
        net_amt = order_amt - ((order_amt/100)*10)
        return net_amt
    else:
        return order_amt



def key(order, order_amt,net_amt):
    if order_amt>100:
        net_amt = order_amt - ((order_amt/100)*5)
        return net_amt
    else:
        return order_amt



def electrical(order, order_amt,net_amt):
    if order_amt>500:
        net_amt = order_amt - ((order_amt/100)*10)
        return net_amt
    else:
        return order_amt


```

```python
def bill(order, order_amt):
    net_amt=0
    if order == int(1):
        net_amt = battery(order, order_amt,net_amt)
        return net_amt
    elif order == int(2) :
        net_amt = key(order, order_amt,net_amt)
        return net_amt
    elif order == int(3):
         net_amt = electrical(order, order_amt,net_amt)
         return net_amt


```

```python
one = "Battery based toys"
two = "Key based toys "
three = "Electrical charging based toys."

print(" we have 3 categories of toys.")
print(f" 1. {one}\n 2. {two}\n 3. {three}\n Type 1,2 and 3")

order = 0
while order != 1 and order != 2 and order != 3:
    order = int(input(f"Type 1 for {one}:\n Type 2 for {two}:\n Type 3 for {three}:"))
    if order !=1 and  order != 2 and order!= 3:
        print("Invalid entry")
        print(f" 1. {one}\n 2. {two}\n 3. {three}\n Type 1,2 and 3")
print(f"You have selected {order}")

order_amt = int(input("Enter order amount:"))

print("Your Net amount is: Rs. ",bill(order, order_amt))
```

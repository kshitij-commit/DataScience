````python
"""Example 5"""

"""Divide the number 148.789 by 12, 13, 14 , 15 separately and add their quotient and remainder."""

a = divmod(148.789,12)
b = divmod(148.789,13)
c = divmod(148.789,14)
d = divmod(148.789,15)

print(a,b,c,d)
````

````python
Q = a[0]+b[0]+c[0]+d[0]  # quotient
R = a[1]+b[1]+c[1]+d[1]  # remainder

R= round(R,2)
print(f"addition of quotient: {Q} \naddition of remainder: {R}")
````

# Variable Scope
where a variable is recognized and can be accessed.

## There are two main types:

1. Local Variable
2. Global Variable

### 1)  Local Variable
A local variable is a variable declared inside a function and can only be used inside that function.

 can only be accessed within that function.

````python
""" Example 1  """

def greet():
    name = "kshitij"   # local variable
    print(name)

greet()
````

### 2) Global variable 
A global variable is declared outside all functions and can be accessed throughout the program.

can be accessed from anywhere in the program.

````python
""" Example 1 """



name = "Kshitij"

def greet():
    print(name)

greet()
print(name)
````

```
Local                       vs          Global
Local Variable	                        Global Variable
Inside function	                        Outside function
Limited scope	                        Wider scope
Accessible only inside function	        Accessible throughout program
```

````python
""" Example 2 """


x = 10

def  show():
    x = 20              # creates a new local variable. It cannot change the global variable.
    print(x)
    

show()
print(x)
````

### 3) global Keyword
Used when we want to modify a global variable inside a function.

````python
""" Example 1 """


x = 10

def show():
    global x
    x = 20
    print(x)

show()
print(x)
print(x)
print(x)
````

# Variable Length Argument in function
Sometimes we don't know how many arguments a user will pass to a function.

```
Example:

Today:

add(10, 20)

Tomorrow:

add(10, 20, 30, 40)
```

## 1. *args (Variable Positional Arguments)

*args allows a function to accept any number of positional arguments.

1. args is stored as a tuple.
2. args are non-keyword argument.

````python
""" Example 1 """


def add(*args):
    print(args)

add(10,20,30)
````

````python
""" Example 2 """


""" Sum of all numbers """

def add(*args):
    sum = 0

    for num in args:
        sum+=num
    return sum
add(10,20,30)
````

```
What is the data type of *args?

Answer:

Tuple
```

## 2. **kwargs (Variable Keyword Arguments)

**kwargs allows a function to accept any number of keyword arguments.

1. kwargs is stored as a dictionary. 
2. Kwargs are keyword argument.

````python
""" Example 1 """


def student(**kwargs):
    print(kwargs)

student(name="Kshitij", branch="cse")
````

```
Question 1. What is the data type of **kwargs?

Answer:

Dictionary
```

````python
""" Example 2 """


def student(**kwargs):
    
    for key, value in kwargs.items():
        print(key, value)

student(name ="Kshitij", branch="cse")
````

````python
""" Example 3 """

def show1(**kwargs):

    for x in kwargs:
        print(kwargs)

  

show1(name = "Kshitij", city =  "Satna")
````

````python
""" Example 4 """


def show(*args, **kwargs):
    print(args)
    print(kwargs)

show(10,20,30, name = "Kshitij", branch = "cse")
````

# Generator function

A generator function is a special function that uses the yield keyword to generate values one at a time.

```
Syntax
def my_generator():
    yield value

```


First Understand the Problem
```
Suppose we want numbers from 1 to 1,000,000.

Normal function:

def numbers():
    result = []

    for i in range(1, 1000001):
        result.append(i)

    return result

```

```
What happens?

Creates all numbers at once
Stores all numbers in memory
Uses lots of RAM

For a million numbers, that's wasteful.
```

Generator Idea

```
Instead of creating everything at once:

Generate values one by one when needed.

Think of it like:

🍽️ Restaurant

Normal Function:

Cook 1000 dishes first.
Then serve.

Generator:

Cook one dish.
Serve.
Cook next dish.
Serve.

Much more efficient.
```

```
What is next()?

next() asks the generator:

"Give me the next value."

```

```
Question 1

How do you identify a generator function?

Answer:

A function containing the yield keyword is a generator function.
```

```
Question 2.

Generator               vs                   Normal Function

Normal Function	                            Generator Function
Uses return	                                Uses yield
Returns all values at once	                Produces values one by one
More memory usage	                        Less memory usage
Ends after return	                        Pauses and resumes




Question 3.

What is the advantage of generators?

They save memory by generating values one at a time.


question 4.

What does next() do?

It retrieves the next value from a generator.


Question 5.

Does a generator use return or yield?

yield

```

```

Now Imagine
def numbers():
    return 1
    return 2
    return 3

Can this return 1, 2, and 3?

No.

Because after the first return:

return 1

the function ends.

So:

return 2
return 3

never execute.

Problem

Suppose I want:

1
2
3
4
5

one by one.

With return, I can't easily do that.

            Enter Yield

yield says:

"Here is a value, but don't destroy the function. Pause it."

Example:

def numbers():
    yield 1
    yield 2
    yield 3

Notice:

No return.

Only yield.

````python
""" Example 1 """


def numbers():
    for i in range(0 ,5):
        yield i

g = numbers()   # this is generator ready 
````

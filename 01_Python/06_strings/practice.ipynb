## 2) String Slicing
Slicing is used to extract a portion (substring) of a string.

Indexing gives one character.

Slicing gives multiple characters.

```
Syntax

string[start:end:step]

1. start is included
2. end is excluded
```

````python
""" Example 1 """

name = "python"

print(name[0:4])
````

### a) Omitting Start
default start by 0 (zero)

````python
""" Example 1 """

text = "hello world"

print(text[:4])
````

### b) Omitting End
default end.

Python goes till the end.

````python
""" Example 1 """

text = "python"
print(text[2:])
````

### c) Negative slicing

Again:

Start included

End excluded

````python
""" Example 1 """

name = "python"

print(name[-4:-1])  # end is excluded
````

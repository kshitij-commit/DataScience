# String

1. String is a data strucure in python that represent a sequence of characters.
2. enclosed within single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """).
3. It is an immutable data type , meaning once you have created a string, you cannot change it. 

Applications:
1. Storing and manipulating text data. 
2. Represent names, addresses.
3. Any text data.

## 1) Indexing in string

Indexing is the process of accessing individual characters of a string using their position number.

````python
""" Example """

text = "Python"
print(f" First character is: {text[0]} and last character is {text[-1]}")
````

````python
""" Example 2 """

""" print only even indexes characters. """


word = "computer"
for i in range(len(word)):
    if (i % 2 == 0):
        print(word[i])
````

## 2) String Slicing
Slicing is used to extract a portion (substring) of a string.

Indexing gives one character.

Slicing gives multiple characters.

```
Syntax

string[start:end:step]

1. start is included
2. end is excluded
```

````python
""" Example 1 """

name = "python"

print(name[0:4])
````

### a) Omitting Start
default start by 0 (zero)

````python
""" Example 1 """

text = "hello world"

print(text[:4])
````

### b) Omitting End
default end.

Python goes till the end.

````python
""" Example 1 """

text = "python"
print(text[2:])
````

### c) Negative slicing

Again:

Start included

End excluded

````python
""" Example 1 """

name = "python"

print(name[-4:-1])  # end is excluded
````

## 1. strip() method

strip() removes unwanted spaces (or specified characters) from the beginning and end of a string.

strip() removes spaces only from:

✅ Beginning

✅ End

Not from the middle.

````python
""" Example 1 """

name = "     python  "

(name)
````

````python
""" Example 2 """


text = "   Python    "
(text.strip())
````

````python
""" Example 3 """

word = "    hello       world.      "

word.strip()
````

### a) removing specific characters


````python
""" Example 1 """

text = "####Python#**###"
print(text.strip("#"))  # middle part didnt remove
````

### b) lstrip()
removes spaces only from the left sides.

````python
""" Example 1 """

name = "        python      "
name.lstrip()

````

### c)  rstrip()
removes spaces only from the right side.

````python
""" Example 1 """

name = "        python      "
name.rstrip()

````

# Slicing 

```python
"Example 1"
name = "programming"
print(name[3:7])
```

```python
""" Example 2 """

num = "123456789"
print(num[::2])
```

```python
""" Example 3 """

word = "coding"

print(word[:-4:-1])
```

```python
""" Example 4 """

""" by using a, b, c and d form a string   apples oranges mangoes are fruit."""

a = "Apples are red in color."
b = "oranges are sweet  and sour in taste. "
c = "mangoes are rich source of vitamins."
d = "These are fruits."

print(f"{a[:6]} {b[:7]} {c[:7]} {d[6:]}")
print(a[:6] +" "+ b[:7] +" "+ c[:7] +" "+ d[6:])

```

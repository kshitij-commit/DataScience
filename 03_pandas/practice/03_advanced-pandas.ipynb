# 3. grouby()

```python
customer_data = {
    "Customer_id":[101,102,103],
    "Name":["Alice","Bob","Charlie"],
    "Age":[25,29,34],
    "Total_Spend":[150.50, 45.00, 300.00],
    "Country":["USA", "CANADA", "USA"]
}
customer_data
```

```python
type(customer_data)
```

```python
customer_data = pd.DataFrame(customer_data)
customer_data
```

```python
type(customer_data)
```

## 3.1 What is the total money made from each country?"

```python
customer_data.groupby("Country")["Total_Spend"].sum()
```

```python
g =customer_data.groupby("Country")
```

```python
g.sum()
```

## 3.2 What is the average age of customers in each country?"

```python
customer_data.groupby("Country")["Age"].mean()
```

```python
g.mean(["Age","Total_Spend"])
```

```python
g.max()
```

# 5. Data Combining & Merging

## 5.1 Concatenate()

```python
import pandas as pd
```

```python
df1 = pd.read_csv("Titanic-Dataset.csv")
df1.head()
```

```python
df2 = df1[["PassengerId","Survived","Pclass","Name"]][0:3]
df2
```

```python
df3 = df1[["PassengerId","Survived","Pclass","Name"]][3:6]
df3
```

### 1. vertical Concate

```python
pd.concat([df2,df3])
```

### 2. horizontal concat

```python
pd.concat([df2,df3],axis=1)
```

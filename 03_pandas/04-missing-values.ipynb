# Topic 1: isnull()


isnull() checks whether values are missing (NaN) or not.

Why Use It?
- Used to identify missing data before cleaning or filling it.

```
Syntax

df.isnull()
df["column_name"].isnull()
```

Important Points

1. Usually combined with sum() to count missing values.

`````python
"Example 1"


df.isnull().sum()
`````

# Topic: 2 notnull()

notnull() checks whether values are present (not missing).

Used to identify valid/non-missing data.

````
Syntax

df.notnull()
df["column_name"].notnull()
```


Important Points

Opposite of isnull().
Returns True for non-missing values.
Returns False for missing values.
Often used for filtering rows.

`````python
df["Lot Frontage"].isnull()  # which values are missing 
`````

`````python
df.columns
`````

`````python
df["Lot Frontage"].notnull()   # which values are not missing/ available.
`````

`````python
df[df['Lot Frontage'].notnull()]  # filter rows where li=ot Frontage exist. 
`````

- count non-missing values.

`````python
df["Lot Frontage"].notnull().sum()
`````

`````python
df.shape
`````

`````python
1460-490
`````

```
isnull() and notnull() are mostly inspection functions.

They help answer:

Where are the missing values?
How many are there?
Which rows have valid data?

But they do not modify the dataset.
```

# Topic 3: dropna()

dropna() removes rows or columns containing missing values.

Why Use It?
- Used when missing data is not useful and can be safely removed.

```
Syntax

df.dropna()
df.dropna(axis=1)
```

Important Points

- Default: removes rows containing at least one missing value.
- axis=0 → rows (default)
- axis=1 → columns
- Returns a new DataFrame unless inplace=True.
- Can remove a large amount of data if used carelessly.

- always runs 

`````python
df.isnull().sum().sort_values(ascending = False).head()
`````

## dropna(axis = 1)

`````python
df.dropna().shape
`````

`````python
df.dropna(axis=1)
`````

`````python
df.dropna(axis=1).shape
`````

## dropna(subset=[...])

`````python
df["SalePrice"].isnull().sum()
`````

`````python
df.dropna(subset=["SalePrice"])  # it will remove 0 rows because no row has missing values.
`````

## 1. how('all') 

Remove rows/columns only if all values are missing.
```
Syntax

df.dropna(how="all")
```
Important Point
- Removes row only when every column in that row is NaN.

## 2. thresh 

Keep rows having at least a specified number of non-missing values.
```
Syntax
df.dropna(thresh=n)
```

Important Point
- thresh counts non-null values, not null values.
 
``` 
df.dropna(thresh=2)

Keep rows having at least 2 non-null values.

Why Use It?
- Used to identify missing data before cleaning or filling it.

```
Syntax

```

Important Points

1. Usually combined with sum() to count missing values.

# Topic: 2 notnull()

notnull() checks whether values are present (not missing).

Used to identify valid/non-missing data.

````
Syntax

df.notnull()
df["column_name"].notnull()
```


Important Points

Opposite of isnull().
Returns True for non-missing values.
Returns False for missing values.
Often used for filtering rows.

# Topic 3: dropna()

dropna() removes rows or columns containing missing values.

Why Use It?
- Used when missing data is not useful and can be safely removed.

```
Syntax

df.dropna()
df.dropna(axis=1)
```

Important Points

- Default: removes rows containing at least one missing value.
- axis=0 → rows (default)
- axis=1 → columns
- Returns a new DataFrame unless inplace=True.
- Can remove a large amount of data if used carelessly.

- always runs 

`````python
df.isnull().sum().sort_values(ascending = False).head()
`````

`````python
df.dropna().shape
`````

`````python
df.dropna(axis=1)
`````

## dropna(subset=[...])

`````python
df.dropna(subset=["SalePrice"])  # it will remove 0 rows because no row has missing values.
`````

## 1. how('all') 

Remove rows/columns only if all values are missing.
```
Syntax

df.dropna(how="all")
```
Important Point
- Removes row only when every column in that row is NaN.

## 2. thresh 

Keep rows having at least a specified number of non-missing values.
```
Syntax
df.dropna(thresh=n)
```

Important Point
- thresh counts non-null values, not null values.
 
```

Keep rows having at least 2 non-null values.

# Topic 4: fillna()

fillna() replaces missing values with a specified value.

Why Use It?
- Used to keep data instead of deleting rows/columns.

```
Syntax

df.fillna(value)
df["column"].fillna(value)
```


Important Points

- Replaces NaN values.
- Can use mean, median, mode, or constant values.
- Usually safer than dropna().
- Choice of value depends on the column.

`````python
df.info()
`````

`````python
df["Lot Frontage"].describe()
`````

### fillna() with Median

Replace missing values with the median of the column.

```
Syntax

df["Lot Frontage"] = df["Lot Frontage"].fillna(
    df["Lot Frontage"].median()
)
```

Why Use Median?

- Good for numeric columns.
- Robust against outliers.
- Commonly used in real datasets.

Important Point

- Median = middle value after sorting.
- Preferred when data may contain extreme values.

`````python
df["Lot Frontage"] = df["Lot Frontage"].fillna(df["Lot Frontage"].median())

df["Lot Frontage"].isnull().sum()
`````

`````python
df["Lot Frontage"]
`````

## fillna() with  Mode 

Ab hum categorical columns par aate hain.

Replace missing values with the most frequently occurring value.

```
Syntax

df["column"] = df["column"].fillna(
    df["column"].mode()[0]
)
```

Why Use It?
- Used for categorical/text columns.

Important Points

- Mode = most frequent value.
- mode() returns a Series.
- Usually use [0] to get the first mode.

### Step 1: Find the Mode

`````python
df["Garage Qual"].value_counts()
`````

`````python
df["Garage Qual"].mode()[0]
`````

### Step 2: Fill Missing Values

`````python
df["Garage Qual"]  = df["Garage Qual"].fillna(df["Garage Qual"].mode()[0])
`````

`````python
df["Garage Qual"].isnull().sum()
`````

## fillna() Strategy 3: Meaningful Missing Value

Replace missing values with a fixed/custom value.

```
Syntax

df["column"] = df["column"].fillna("value")
```
Used when missing values have a meaningful interpretation.

Important Points

- Common for categorical columns.
- Preserves information.
- Often better than mode for certain columns.


`````python
(df["Pool QC"])
`````

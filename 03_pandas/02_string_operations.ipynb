# 1. String Data Type

## Why?

Pandas needs a datatype for storing text data.

Earlier strings were stored using object dtype.

Now pandas recommends str/string dtype.

In older Pandas versions, text columns are often stored as:

dtype: object

In newer Pandas versions (3.x), Pandas can infer:

dtype: str

To explicitly create a string column:

dtype="string"
## Key Point

dtype="str" means the column contains text data.

````python
import pandas as pd

print(pd.__version__)
````

````python
import pandas as pd


"Example 1"
s = pd.Series(
    ["Ram", "Shyam", "Mohan"],
    dtype="string"
)

print(s)
print(s.dtype)
````

````python
"Example 2"

s = pd.Series([
    "apple",
    "banan"],
    # dtype="string"
)
print(s.dtype)
````

````python
import pandas as pd

print(pd.__version__)

s = pd.Series(
    ["Ram", "Shyam", "Mohan"],
    dtype="string"
)

print(s.dtype)
````

# 2. .str Accessor

## Why?

A Pandas Series is not a Python string.

Therefore string methods like upper(), lower(), strip() cannot be called directly on a Series.

## Solution

Use .str before string methods.



````python
'Example 1'
s = pd.Series(
    ["ram", "shyam", "mohan"],
    dtype="string"
)

s.str.upper()
````

## a) .str.len()

Returns the length of each string in a Series.

````python
" Example: 1"

s = pd.Series(
    ["Ram", "Shyam", "Mohan"],
    dtype="string"
)

s.str.len()
````

## Method Chaining

Multiple operations can be performed in one line.

Example:

s.str.upper().str.len()

Execution:
```
Series
 → upper()
 → len()
 → Result
 ```

## b) .str.strip()

Removes leading and trailing spaces from every string in a Series.

Related Methods:

.str.lstrip() -> remove left spaces

.str.rstrip() -> remove right spaces


````python
"Example:"

s.str.strip()
````

````python
"Example 2"

s = pd.Series(
    ["ram@gmail.com",
     "shyam@yahoo.com"],
    dtype="string"
)
s.str.split('@')
````

## c) .str.split()

Splits each string using a separator.

Syntax:

series.str.split(separator)

Example:

- "ram@gmail.com"
- → ["ram", "gmail.com"]

## d) .str.get()

Used to access an element after split().

Example:

[ram, gmail.com]

get(0) → ram

get(1) → gmail.com

Shortcut:
```
 .str.get(1) = .str[1]
```

````python
"Example 1"

s = pd.Series(
    ["A-B-C",
     "X-Y-Z"],
    dtype="string"
)

s.str.split("-").str.get(2)
````

## d) .str.split(..., expand=True)

Converts split output into separate columns.
```
Example:

"ram@gmail.com"

↓

ram | gmail.com
```
Syntax:

series.str.split("@", expand=True)
```
Without expand=True:

[A, B, C]

With expand=True:

A | B | C

Result becomes a DataFrame.
```

````python
"Example 1"

s = pd.Series(
    ["A-B-C",
     "X-Y-Z"],
    dtype="string"
)

s.str.split("-", expand=True)
````

## e) n Parameter

Controls the maximum number of splits.

Example:

"A-B-C"

split("-", n=1)

Result:

["A", "B-C"]

````python
"Example 1"

s = pd.Series(
    ["python-data-science"],
    dtype="string"
)

s.str.split("-", n=1)
````

## f) .str.rsplit()

Splits from the right side.

Example:

"A-B-C"
```
rsplit("-", n=1)
```
Result:

["A-B", "C"]

````python
"Example 1"

s = pd.Series(
    ["machine-learning-python"],
    dtype="string"
)

s.str.rsplit("-", n=1)
````

## k) .str.extract()

Extracts information from strings using capture groups.

Each capture group becomes a column.

```
Syntax:

series.str.extract(pattern)
```

```
Example:- s.str.extract(r"([A-Z]+)(\d+)")

Input:

AB123

Output:

AB | 123
```

```
Pattern     	Meaning
[A-Z]	        One capital letter
[A-Z]+	        One or more capital letters
\d	            One digit
\d+	            One or more digits

```

````python
"Example 1"

s = pd.Series(
    "abc123xyz"
)
s.str.findall(r"\d")

````

````python
"Example 1"

s = pd.Series([
    "A1",
    "B2",
    "C3"
])

s.str.extract(r"([A-Z])(\d)")
````

````python
"Example 2"

s = pd.Series([
    "X7",
    "Y8A",
    "Z9"
])

s.str.extract(r"([A-Z])(\d)")
````

````python
"Example 3"

s = pd.Series([
    "CS101",
    "ME205",
    "EE301"
])

s.str.extract(r"([A-Z]+)([\d]+)")
````

````python
"Example 4"

s = pd.Series([
    "CS-101",
    "ME-205",
    "EE-301"
])

s.str.extract(r"([A-Z]+)-(\d+)")
````

## l) .str.extractall()

Returns every match.

```
Example: "x7y8"

extract()

Result:  x | 7

extractall()

Result:  x | 7
         y | 8
```


Difference:

extract()    -> first match

extractall() -> all matches

````python
"Example 1"

s = pd.Series([
    "x7y8"
])

s.str.extractall(r"([a-z])(\d)")
````

## Q) Regex

### Regex: \w

Represents a word character.

Matches:

Letters (a-z, A-Z)
Digits (0-9)
Underscore (_)

Does NOT match:

@
#
$
space

````python
s = pd.Series([
    "A_1",
    "#B2"
])

s.str.findall(r"\w")
````

### Regex: \s

Represents whitespace.

Matches:

space
tab
newline

Examples:

"AI ML"

count(r"\s")

Result:

1

````python
s = pd.Series([
    "Data Science",
    "Machine Learning"
])

s.str.findall(r"\s")
````

````python
"Example 1"

s = pd.Series([
    "color",
    "colour",
    "colouur"
])

s.str.fullmatch(r"colou?r")
````

### Regex: ?

Means "zero or one occurrence".

Also called:

Optional

Example:

colou?r

Matches:

color
colour

Does NOT match:

colouur

Reason:

u? allows:

0 occurrence of u
or
1 occurrence of u

but not 2 or more.

````python
s = pd.Series([
    "color",
    "colour",
    "colouur"
])

s.str.fullmatch(r"colou?r")
````

```
\d      -> Digit (0-9)

[A-Z]   -> One capital letter

+       -> One or more occurrences

\w      -> Letter, Digit, Underscore

\s      -> Whitespace

?       -> Zero or one occurrence (Optional)

```

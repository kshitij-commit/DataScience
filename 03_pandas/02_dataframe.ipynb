# 1. What is a DataFrame?

DataFrame is a 2-dimensional labeled data structure.

Think:

Series = One Column

DataFrame = Multiple Columns

A DataFrame can be viewed as:

1. Spreadsheet
2. SQL Table
3. Dictionary of Series

Rows -> Records

Columns -> Features

# 2. DataFrame Creation: From Dict of Series

Dictionary keys become column names.

Each Series becomes a column.

Pandas aligns Series using index labels.

Missing values become NaN.

````python
import pandas as pd
````

````python
"Example 1"


d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}
print(d)
````

````python
df = pd.DataFrame(d)
print(df)
````

## 2.1 Custom Index            

index parameter controls row labels.
```
Example:

pd.DataFrame(
    d,
    index=["d","b","a"]
)
```

```    
    Only rows d, b, a are kept.

    Rows not listed are removed.

    Missing values become NaN.
    ```

````python
"Example 1"


d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}

````

````python
pd.DataFrame(
    d,
    index=['b','a','d']
)
````

## 2.2 Custom Columns

columns parameter controls column labels.
```
Example:

pd.DataFrame(
    d,
    columns=["two","three"]
)

```

```
    Column "two" exists.

    Column "three" does not exist.

    Result:

    Column "three" is filled with NaN.

```

````python
"Example 1"



d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}

pd.DataFrame(
    d,
    index=['d','b','a'],
    columns=['two','three']
)

````

# 3. DataFrame from Dict of Lists

Dictionary keys become column names.

Lists become column values.

All lists must have the same length.

- Default index:0,1,2,...

````python
"Example 1"


d = {
    'one': [1,2,3,4],
    'two': [2,3,4,5]
}

pd.DataFrame(d)
````

### Custom Index with Dict of Lists

Use index= to specify row labels.

Example:

pd.DataFrame(
    data,
    index=["A","B","C"]
)

Rules:

1. Length of index must equal number of rows.
2. Data remains unchanged.
3. Only row labels are modified.

````python
"Example 2"

d = {
    "one": [1.0,2.0,3.0,4.0],
    "two": [4.0,3.0,2.0,1.0]
}

pd.DataFrame(
    d,
    index=["a","b","c","d"]
)
````

# 4. List of dictionaries

1. DataFrame from List of Dicts
2. Each dictionary represents one row.
3. Dictionary keys become column names.
4. Dictionary values become row values.
5. Missing keys are filled with NaN.
6. Commonly used when working with JSON, APIs, and web scraping data.

```
Pandas takes the union of all keys from all dictionaries.
```

````python
import pandas as pd
````

````python
"Example 1"

s= [
    {
        'one':1,
        'two':2
    },
    {
        'hello':"world",
        'data':'science'
    }
]

pd.DataFrame(s)
````

````python
"Example 1"


data = [
    {"name":"Ram","age":20},
    {"name":"Shyam"},
    {"name":"Mohan","age":22}
]

df = pd.DataFrame(data)
df
````

# 5. Creating a DataFrame from a Series

DataFrame from Series
1. A Series can be converted into a DataFrame.
2. The resulting DataFrame contains one column.
3. Series index becomes DataFrame index.
4. Series name becomes DataFrame column name

````python
"Example 1"

import pandas as pd

ser = pd.Series(
    [10,20,30],
    # name="Marks"
)

df = pd.DataFrame(ser)

print(df)
````

# 7. Adding New Columns

In Data Science, we often create new features from existing columns.

Adding New Columns
```
Syntax:

df["new_column"] = value
```
Possible values:

- Scalar
- List
- Series
- Calculation from existing columns

## 7.1 Method 1: Add a Constant Value

````python
"Example 1"

df = pd.DataFrame({
    "Marks":[20,30,40,50]
})
print(type(df))
print((df))
````

# 8. Deleting Columns

## 8.1 Method 1: del

1. The column is permanently removed from the DataFrame.

2. Deletes the column.  You don't get the deleted column back.

```
Syntax:

del df["column_name"]
```

````python
"Example 1"

df = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})

del df["B"]

print(df)
````

## 8.2 Method 2: pop()

1. Deletes and returns the columns as Series..

```
Syntax:

df.pop("column_name")
```

````python
"Example 1"

df = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})

x = df.pop("B")

print(df)
````

````python
print(x)
print(type(x))
````

# 9. df.insert()

Used to insert a column at a specific position.

```

Syntax
df.insert(
    loc,
    column_name,
    values
)
```
Meaning
1. loc → Position where column should be inserted
2. column_name → Name of new column
3. values → Data for column



````python
"Example 1"


df = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})

df.insert(0, "X", [10,20])

print(df)
````

````python
"Example 2"

df = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})

df.insert(1, "C", [5,6])
df
````

# 10. Row Selection with loc

Used for label-based row selection.

Syntax:

df.loc["label"]

Example:

df.loc["b"]

Returns:
```
Name     Shyam
Marks       90

Uses row labels.
Returns a Series for a single row.
Raises KeyError if label does not exist.

```

````python
"Example 1"

df = pd.DataFrame(
    {
        "Name":["Ram","Shyam","Mohan"],
        "Marks":[80,90,70]
    },
    index=["a","b","c"]
)

df.loc["b"]
````

## 10.1 single row 

````python
"Example 2 "


df = pd.DataFrame(
    {
        "City":["Delhi","Mumbai","Pune"],
        "Population":[30,20,10]
    },
    index=["x","y","z"]
)

print(df.loc["z"])
````

## 10.2 multiple rows

````python
"Example 1"

df = pd.DataFrame(
    {
        "Name":["Ram","Shyam","Mohan"],
        "Marks":[80,90,70]
    },
    index=["a","b","c"]
)
print(df.loc[["a","c"]])
print(type(df.loc[["a","c"]]))
````

## 10.3 Single Cell  ->  loc[row, column]

Returns single value only.

````python
"Example 1"

df = pd.DataFrame(
    {
        "City":["Delhi","Mumbai","Pune"],
        "Population":[30,20,10]
    },
    index=["x","y","z"]
)

df.loc["z", "Population"]
````

## 10.4 Selecting Rows and Columns Together

Used to select specific rows and specific columns.

```
syntax: df.loc[[rows],[columns]]

df.loc[rows, columns]


df.loc[
    which rows,
    which columns
]
```

````python
"Example 1"

df = pd.DataFrame(
    {
        "Name":["Ram","Shyam","Mohan"],
        "Marks":[80,90,70],
        "Age":[20,21,19]
    },
    index=["a","b","c"]
)

print(df.loc[['a','c'],["Name","Age"]])
df.loc[['a'],["Name"]]
````

## 10.5 Label Slicing

Includes both start and end labels.
 


````python
df = pd.DataFrame(
    {"Marks":[80,90,70]},
    index=["a","b","c"]
)
df.loc["a":"c"]
````

# 11. iloc (Integer Location)

iloc only accepts integer positions.
- - Used for position-based selection.

````python
"Example 1"

df = pd.DataFrame(
    {
        "City":["Delhi","Mumbai","Pune"]
    },
    index=["x","y","z"]
)

print(df.iloc[2])
````

## 11.1 Multiple Rows with iloc

 - df.iloc[rows, columns]

````python
"Example 1"

df = pd.DataFrame(
    {
        "City":["Delhi","Mumbai","Pune"]
    },
    index=['x','y','z']
)

print(df.iloc[[1,2]])
print(type(df.iloc[[1,2]]))
````

````python
df.iloc["y"]  ## Gives error 
````

## 11.2 iloc[row_position, column_position]

```
df.loc[row_label, column_label]

- Returns a single cell

```

````python
"Example 1"

df = pd.DataFrame(
    {
        "Name":["Ram","Shyam","Mohan"],
        "Marks":[80,90,70]
    }
)


print(df.iloc[0,1])
# print(df.iloc[[0,1],0])
df.iloc[0,0]
````

##  11.3 multiple_row and multiple_columns 

df.iloc[row_positions, column_positions]

````python
"Example 1"

df = pd.DataFrame(
    {
        "Name":["Ram","Shyam","Mohan"],
        "Marks":[80,90,70]
    }
)

print(df.iloc[[0,1],0])
````

# 1. What is a DataFrame?

DataFrame is a 2-dimensional labeled data structure.

Think:

Series = One Column

DataFrame = Multiple Columns

A DataFrame can be viewed as:

1. Spreadsheet
2. SQL Table
3. Dictionary of Series

Rows -> Records

Columns -> Features

# 2. DataFrame Creation: From Dict of Series

Dictionary keys become column names.

Each Series becomes a column.

Pandas aligns Series using index labels.

Missing values become NaN.

````python
import pandas as pd
````

````python
"Example 1"


d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}
print(d)
````

````python
df = pd.DataFrame(d)
print(df)
````

## 2.1 Custom Index            

index parameter controls row labels.
```
Example:

pd.DataFrame(
    d,
    index=["d","b","a"]
)
```

```    
    Only rows d, b, a are kept.

    Rows not listed are removed.

    Missing values become NaN.
    ```

````python
"Example 1"


d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}

````

````python
pd.DataFrame(
    d,
    index=['b','a','d']
)
````

## 2.2 Custom Columns

columns parameter controls column labels.
```
Example:

pd.DataFrame(
    d,
    columns=["two","three"]
)

```

```
    Column "two" exists.

    Column "three" does not exist.

    Result:

    Column "three" is filled with NaN.

```

````python
"Example 1"



d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}

pd.DataFrame(
    d,
    index=['d','b','a'],
    columns=['two','three']
)

````

- df.columns -> Returns column labels

# 3. DataFrame from Dict of Lists

Dictionary keys become column names.

Lists become column values.

All lists must have the same length.

- Default index:0,1,2,...

````python
"Example 1"


d = {
    'one': [1,2,3,4],
    'two': [2,3,4,5]
}

pd.DataFrame(d)
````

### Custom Index with Dict of Lists

Use index= to specify row labels.

Example:

pd.DataFrame(
    data,
    index=["A","B","C"]
)

Rules:

1. Length of index must equal number of rows.
2. Data remains unchanged.
3. Only row labels are modified.

````python
"Example 2"

d = {
    "one": [1.0,2.0,3.0,4.0],
    "two": [4.0,3.0,2.0,1.0]
}

pd.DataFrame(
    d,
    index=["a","b","c","d"]
)
````

# 4. List of dictionaries

1. DataFrame from List of Dicts
2. Each dictionary represents one row.
3. Dictionary keys become column names.
4. Dictionary values become row values.
5. Missing keys are filled with NaN.
6. Commonly used when working with JSON, APIs, and web scraping data.

```
Pandas takes the union of all keys from all dictionaries.
```

````python
import pandas as pd
````

````python
"Example 1"

s= [
    {
        'one':1,
        'two':2
    },
    {
        'hello':"world",
        'data':'science'
    }
]

pd.DataFrame(s)
````

````python
"Example 1"


data = [
    {"name":"Ram","age":20},
    {"name":"Shyam"},
    {"name":"Mohan","age":22}
]

df = pd.DataFrame(data)
df
````

# 5. Creating a DataFrame from a Series

DataFrame from Series
1. A Series can be converted into a DataFrame.
2. The resulting DataFrame contains one column.
3. Series index becomes DataFrame index.
4. Series name becomes DataFrame column name

````python
"Example 1"

import pandas as pd

ser = pd.Series(
    [10,20,30],
    # name="Marks"
)

df = pd.DataFrame(ser)

print(df)
````

````python
"Example 2"

ser = pd.Series(
    [85,90,78],
    index=["Ram","Shyam","Mohan"],
    name="Marks"
)

pd.DataFrame(ser)
````

# 6. Column Selection (df["column"])

Column Selection
```
Syntax:

df["column_name"]
```
- Used to select a single column.
- Returns a Series.
- Column name must exist.
- Non-existing column raises KeyError.

````python
df = pd.DataFrame({
    "Name":["Ram","Shyam","Mohan"],
    "Marks":[85,90,78]
})
df
````

````python
print(type(df["Marks"]))
df["Marks"]
````

````python
df["Marks"].mean()
````

## 6.1 Multiple Column Selection

Syntax:

- df[["col1","col2"]]
- Used to select multiple columns.
- Returns a DataFrame.
- Column names are passed inside a list.

````python
"Example 1"

import pandas as pd

df = pd.DataFrame({
    "Name":["Ram","Shyam","Mohan"],
    "Age":[20,21,19],
    "Marks":[85,90,78]
})


````

````python
print(df[['Name','Marks']])
print(type(df[['Name','Marks']]))
````

````python
print(df["Marks"])
print(type(df["Marks"]))
````

````python
print(df[["Marks"]])
print(type(df[["Marks"]]))
````

# 7. Adding New Columns

In Data Science, we often create new features from existing columns.

Adding New Columns
```
Syntax:

df["new_column"] = value
```
Possible values:

- Scalar
- List
- Series
- Calculation from existing columns

## 7.1 Method 1: Add a Constant Value

````python
"Example 1"

df = pd.DataFrame({
    "Marks":[20,30,40,50]
})
print(type(df))
print((df))
````

````python
df['Age'] = 18

print(df)
print(type(df))

````

## Method 2: Create Column from Existing Columns

````python
"Example 1"

df = pd.DataFrame({
    "Marks":[80,90,70],
    "Bonus":[5,5,5]
})
df["Total"]  =  df["Marks"] + df["Bonus"]

print(type(df))
print((df))
````

## Method 3: Add Using a List

````python
"Example 1"


df = pd.DataFrame({
    "Marks":[80,90,70],
    "Bonus":[5,5,5]
})


df["Grade"] = ['A','B',"C"]

````

````python
print(type(df))
print((df))
````

# 8. Deleting Columns

## 8.1 Method 1: del

1. The column is permanently removed from the DataFrame.

2. Deletes the column.  You don't get the deleted column back.

```
Syntax:

del df["column_name"]
```

````python
"Example 1"

df = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})

del df["B"]

print(df)
````

## 8.2 Method 2: pop()

1. Deletes and returns the columns as Series..

```
Syntax:

df.pop("column_name")
```

````python
"Example 1"

df = pd.DataFrame({
    "A":[1,2],
    "B":[3,4]
})

x = df.pop("B")

print(df)
````

````python
print(x)
print(type(x))
````

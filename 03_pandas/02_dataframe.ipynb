# 1. What is a DataFrame?

DataFrame is a 2-dimensional labeled data structure.

Think:

Series = One Column

DataFrame = Multiple Columns

A DataFrame can be viewed as:

1. Spreadsheet
2. SQL Table
3. Dictionary of Series

Rows -> Records

Columns -> Features

# 2. DataFrame Creation: From Dict of Series

Dictionary keys become column names.

Each Series becomes a column.

Pandas aligns Series using index labels.

Missing values become NaN.

````python
import pandas as pd
````

````python
"Example 1"


d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}
print(d)
````

````python
df = pd.DataFrame(d)
print(df)
````

## 2.1 Custom Index            

index parameter controls row labels.
```
Example:

pd.DataFrame(
    d,
    index=["d","b","a"]
)
```

```    
    Only rows d, b, a are kept.

    Rows not listed are removed.

    Missing values become NaN.
    ```

````python
"Example 1"


d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}

````

````python
pd.DataFrame(
    d,
    index=['b','a','d']
)
````

## 2.2 Custom Columns

columns parameter controls column labels.
```
Example:

pd.DataFrame(
    d,
    columns=["two","three"]
)

```

```
    Column "two" exists.

    Column "three" does not exist.

    Result:

    Column "three" is filled with NaN.

```

````python
"Example 1"



d = {
    "one":pd.Series(
        [10,20,30],
        index=['a','b','c']
    ),
    "two":pd.Series(
        [10,20,30,40],
        index=['a','b','c','d']
    )
}

pd.DataFrame(
    d,
    index=['d','b','a'],
    columns=['two','three']
)

````

# 3. DataFrame from Dict of Lists

Dictionary keys become column names.

Lists become column values.

All lists must have the same length.

- Default index:0,1,2,...

````python
"Example 1"


d = {
    'one': [1,2,3,4],
    'two': [2,3,4,5]
}

pd.DataFrame(d)
````

### Custom Index with Dict of Lists

Use index= to specify row labels.

Example:

pd.DataFrame(
    data,
    index=["A","B","C"]
)

Rules:

1. Length of index must equal number of rows.
2. Data remains unchanged.
3. Only row labels are modified.

````python
"Example 2"

d = {
    "one": [1.0,2.0,3.0,4.0],
    "two": [4.0,3.0,2.0,1.0]
}

pd.DataFrame(
    d,
    index=["a","b","c","d"]
)
````

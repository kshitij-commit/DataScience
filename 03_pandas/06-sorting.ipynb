# Topic 1: sort_values()

sort_values() is used to sort data based on column values.


Used to find:

- Highest values
- Lowest values
- Latest records
- Top performers

```
Syntax

df.sort_values("column_name")
```
Important Points

- Default: Ascending order (small → large, A → Z)
- Returns a new DataFrame.
- Original DataFrame remains unchanged unless assigned back.
- One of the most frequently used Pandas functions.

````python
import pandas as pd 
````

````python
df = pd.read_csv("netflix_titles.csv")
df.head(4)
````

````python
df.shape
````

````python
df["release_year"].sort_values()
````

````python
df.sort_values("release_year")[["title","release_year"]].head(20)
````

````python
df["release_year"].sort_values(ascending=False)
````

````python
df.sort_values(
    "release_year",
    ascending=False
)[["title","release_year"]].head(10)
````

## Multiple Column Sorting

Sort data using more than one column.
```
Syntax

df.sort_values(
    ["col1", "col2"]
)
```
Important Points

- First column = primary sorting key
- Second column = tie-breaker
- Third column = tie-breaker for second column
- Very common in real datasets

```

df.sort_values(
    ["release_year", "title"]
)
```

Step 1: Sort by release_year
- 2020
- 2021
- 2021

Step 2: For rows having same year
- Now Pandas uses the second column: title
- and sorts alphabetically:

````python
"Example 1"


df.sort_values(['release_year','title'])[['title',"release_year"]].head(10)
````

--------------------------------------------------
```
df.sort_values(
    ["release_year", "title"],
    ascending=[False, True]
)

- release_year → False
- title → True
```
---------------------------------------------------

````python
"Example 2"

df.sort_values(
    ["release_year", "title"],
    ascending=[False, True]
).head(5)
````

# Topic 2: sort_index()

Sorts a DataFrame by its index labels.

Why Use It?
Used when row order/index has become messy and needs to be arranged.
```
Syntax

df.sort_index()
```
Important Points

- Sorts index, not column values.
- Default = Ascending index order.
- Mostly useful after filtering, sorting, concatenation, etc.

````python
df.sort_values(
    "release_year"
)[["title","release_year"]].head(10)
````

````python
df.sort_values(
    "release_year"
)[["title","release_year"]].head(10).sort_index()
````

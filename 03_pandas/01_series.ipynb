# 9. Automatic Label Alignment

Operations between Series happen by index labels.

- Match labels first.
- Then do the operation.

1. Series operations use labels, not positions.

2. Pandas first aligns indexes.

3. Matching labels are operated on.

4. Missing labels become NaN.



```python
"Example 1"

s1 = pd.Series(
    [1,2,3,4],
    index=['a','b','c','d']

)

s2 = pd.Series(
    [10,20,30,40],
    index=['b','a','d','c']
)

print(s1+s2)
```

```python
"Example 2"

s1 = pd.Series(
    [1,2,3,4],
    index=['a','b','c','d']

)

s2 = pd.Series(
    [10,20,30,40],
    index=['b','a','e','f']
)

print(s1+s2)
```

```python
"Example 3"


s1 = pd.Series(
    [1,2,3,4],
    # index=['a','b','c','d']

)

s2 = pd.Series(
    [10,20,30,40],
    # index=['b','a','e','f']
)

s1+s2
```

```python
"Example 4"


s1 = pd.Series(
    [1,2,3,4],
    index=['a','b','c','d']

)

s2 = pd.Series(
    [10,20,30,40],
    # index=['b','a','e','f']
)

s1 + s2
```

# 1. What is a Series?

Series is a one dimensional labeled array capable of holding any data type.  

- Data + Labels
- Series = Data + Index

```python


import pandas as pd


```

```python
"Example 1"


s = pd.Series([10,20,30])
s
```

```python
"Example 2"

s = pd.Series(
    [10,20,30],index=['a','b','c']
)

s
```

Notice:

a
b
c

are labels.

Pandas calls these labels:

Series = Data + Index

# 2. Creating Series

Series can be created from:

1. ndarray
2. dictionary
3. scalar value

## 2.1 list/ndarray 

### a) default index

```python
"Example 1"

s = pd.Series(
    [10,20,30]
)

print(s)
print(type(s))
```

### b) Custom Index

```python
"Example 1"

s = pd.Series(
    [10,20,30],
    index= ['@','#','%']

)

print(s)
print(type(s))
```

## 2.2 From Dictionary

- Keys -> Index
- Values -> Data

```python
"Example 1"
d = {
    "a":100,
    "b":200,
    "c":300
}

s = pd.Series(d)
print(s)
print(type(s))
```

## 2.3 Scalar

```python
"Example 1"

pd.Series(
    5,
    index=["a","b","c"]
)
```

# What is a Series?

Series is a one dimensional labeled array capable of holding any data type.  

- Data + Labels
- Series = Data + Index

```python


import pandas as pd


```

```python
"Example 1"


s = pd.Series([10,20,30])
s
```

```python
"Example 2"

s = pd.Series(
    [10,20,30],index=['a','b','c']
)

s
```

Notice:

a
b
c

are labels.

Pandas calls these labels:

Series = Data + Index

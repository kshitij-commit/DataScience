# Topic 1: duplicated()

duplicated() checks whether a row is a duplicate of a previous row.

Why Use It?
- Used to identify repeated records in a dataset.

```
Syntax

df.duplicated()
df.duplicated().sum()
```
Important Points

- Returns True for duplicate rows.
- Returns False for first occurrence.
- Does not remove duplicates.
- Usually used before drop_duplicates().

## Duplicate based on specific columns

Suppose we want to check whether the same property appears multiple times.

```
We can do:

df["PID"].duplicated().sum()

or

df.duplicated(subset=["PID"]).sum()

This checks duplicates only on the PID column.
```

````python
df["show_id"].duplicated().sum()
````

# Topic 2: drop_duplicates()

Removes duplicate rows from a DataFrame.

Why Use It?
- Used to eliminate repeated records and prevent incorrect analysis.


```
Syntax

df.drop_duplicates()
df.drop_duplicates(subset=["column"])
df.drop_duplicates(inplace=True)


```

Important Points

- Keeps the first occurrence by default.
- Returns a new DataFrame unless inplace=True.
- Can remove duplicates based on specific columns using subset.
- Always inspect duplicates before deleting them.

## Important Parameter

### 1) keep

Default:

```
df.drop_duplicates(keep="first")
- Keep first occurrence.

Other options:
df.drop_duplicates(keep="last")
- Keep last occurrence.

df.drop_duplicates(keep=False)
- Remove all duplicated records.
```

# Topic 3: rename()

rename() is used to change row labels or column names.

Why Use It?
- Makes column names more readable and meaningful.

```
Syntax

df.rename(columns={
    "old_name": "new_name"
})
```
Important Points

- Original DataFrame is unchanged unless assigned back or inplace=True.
- Can rename one or multiple columns.
- Mostly used for column names.

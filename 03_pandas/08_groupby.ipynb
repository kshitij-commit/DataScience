# GroupBy

groupby() splits data into groups based on a column and allows aggregate calculations on each group.

Why Use It?
- Used to summarize data by categories.

```
Common Operations

mean()
sum()
count()
max()
min()
size()

Syntax

df.groupby("column")
```

````python
import pandas as pd
````

````python
df = pd.read_csv("netflix_titles.csv")
df.head(3)
````

````python
df.groupby("type")
````

````python
df.groupby("type").size()
````

````python
df["type"].value_counts()
````

````python
df["type"].max()
````

## Aggregation

````python
df.groupby('type')['release_year'].mean()
````

## Multiple Aggregations

agg() applies multiple functions at once.

````python
df.groupby("type")["release_year"].agg(
    ['mean','min','max']
)
````

## 1. GroupBy + Count

````python
df.groupby('type')["release_year"].count()
````

## 2. GroupBy + Multiple Columns

````python
df.head(3)
````

````python
df.groupby('country')
````

````python
df.groupby(['type','rating']).count()
````

````python
df.groupby('type').count()
````

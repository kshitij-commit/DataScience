## 1. Parameter : nrows

Load only first N rows.

````python
pd.read_csv("Titanic-Dataset.csv",nrows=100)
````

## 2. usecols

Load selected columns

````python
pd.read_csv(
    "Titanic-Dataset.csv",
    usecols=["Survived","Pclass"]
)
````

## 3. index_cols

Set columns as index

````python
pd.read_csv(
    "Titanic-Dataset.csv",
    index_col=["Pclass"]
)
````

##  4. skiprows 

Skip rows.

````python
pd.read_csv(
    "Titanic-Dataset.csv",
    skiprows=1
)
````

## 5. navalues

Custom missing values.

Convert all into NaN.


```
df = pd.read_csv(
    "file.csv",
    na_values=["NA","Missing","Null"]
)
```

````python
pd.read_csv(
    "Titanic-Dataset.csv",
    na_values=["NaN","missing"]
)
````

## Topic 3: isin()

1. Check whether values belong to a list.

```
Syntax
df["column"].isin([...])
```

Suppose you want: Passengers whose Pclass is: 1 OR 2

```
You could write:

df[
    (df["Pclass"] == 1)
    |
    (df["Pclass"] == 2)
]
```
Works.
But now imagine:

Pclass = 1 OR 2 OR 3 OR 4 OR 5

Now your code starts looking like a conspiracy wall.

Solution: isin()
```
df["Pclass"].isin([1, 2])

Output:

True
True
False
True
...
```
A Boolean Series.

````python
"Example 1"

df[df["Pclass"].isin([1,2])]
````

````python
"Example 2"


df[df["Embarked"].isin(["C"])]
````

### using with loc

````python
"Example 3"

df.loc[
    df["Embarked"].isin(["C"]),
    ["Name"]

]
````

### Not in

````python
df[
    ~(df["Pclass"].isin([3]))
]
````

# Topic 5: value_counts()

- Count frequency of unique values.


Many columns contain categories.

Examples:

- Sex
- Pclass
- Embarked
- Department
- City
- Country
- Product Category

Analysts constantly want to know:

How many times does each category occur?

That's exactly what value_counts() answers.

```
Syntax
df["column"].value_counts()
```

```
Return Type -> pandas.Series
```



df["Sex"].value_counts()

will return-

- male      577
- female    314

Because Pandas sorts by count descending by default.

````python
import pandas as pd
````

````python
df = pd.read_csv("Titanic-Dataset.csv")

df.head(3)
````

````python
"Example 1"

df["Pclass"].value_counts()
````

````python
"Example 2"

df["Sex"].value_counts()
````

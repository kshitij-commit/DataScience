How spread out are values?

```
Very close together.

```

- Small std → values clustered
- Large std → values scattered

### 5.4. Quartiles

Imagine sorted ages:
```
10 15 18 20 22 25 28 30 35 40 50
25% ---------- 50% --------- 75%
```
These values help understand distribution.



#### a) 25%
25% of observations are below this value.

```
Example:

Age 25% = 20

Meaning:

25% passengers are younger than 20.
```

#### b) 50%

Also called:

Median

```
Example:

Age 50% = 28

Meaning:

Half of passengers are younger than 28.
Half are older than 28.
```

#### c) 75%

Example:

Age 75% = 38

Meaning:

75% passengers are younger than 38.

````python
import pandas as pd 
````

````python
df = pd.read_csv("Titanic-Dataset.csv")
````

# Question 1. 

```
Q1
Show:

Female passengers
AND
Age > 30

Return:

Name
Age

using loc.
```

````python
df.loc[

    ((df["Sex"] == "female")

      &

      (df["Age"] > 30) ),

       ["Name","Age"]



]
````

# Questions 2 

```
Q2

Show passengers whose:

Pclass is 1 or 2

Return:

Name
Pclass

using isin() and loc.
```

````python
df.loc[
    (df["Pclass"].isin([1,2])),
    ["Name","Pclass"]
]
````

# Question 3

```
Q3

Show passengers whose:

Fare between 50 and 100
AND
Survived = 1

Return:

Name
Fare
```

````python
df.loc[
    ((df["Survived"] == 1)
    &
    (df["Fare"].between(50, 100))),
    ["Name","Fare"]
]
````

# 1. Introduction to Numpy

## 1.1 What is NumPy?

NumPy = Numerical Python

It is a Python library designed for:

1. Fast numerical computation
2. Working with arrays and matrices
3. Mathematical operations
4. Scientific computing

### Homogeneous Array

A NumPy array is homogeneous, which means all elements are stored as the same data type.

Example:

np.array([1,2,3])
- int64

np.array([1,2,3.5])
- float64

NumPy automatically converts data to a common type when needed.

## 1.2 Why was numpy created 

Imagine storing:

[10, 20, 30, 40]

in a Python list.

A Python list does not store the actual integers directly in a compact block.

Instead, it stores references (addresses) to Python objects.

Simplified:
```
    List
    │
    ├──► Integer Object(10)
    ├──► Integer Object(20)
    ├──► Integer Object(30)
    └──► Integer Object(40)
```
This makes python list flexible:- 
1. Can store int
2. float
3. string
4. list
5. anything

But:

❌ More memory

❌ Slower numerical computation

NumPy says:

"If all data is of the same type, why waste memory?"

So NumPy stores data in a continuous memory block.

Simplified:

[10][20][30][40]

Benefits:

✅ Less memory

✅ Faster access

✅ Vectorized operations

✅ Used by Pandas, Scikit-Learn, TensorFlow, PyTorch

## 1.3 What Problem Does NumPy Solve?



### 1.3.1 Python List vs Numpy array 

Python List Problems
1. Slower computation
2. More memory usage
3. Not optimized for mathematical operations
4. Poor support for matrix operations

NumPy Solution
1. Faster computation
2. Less memory
3. Vectorized operations
4. Matrix and linear algebra support

### 1.3.2 What is Vectorization?

Performing operations on an entire array at once instead of processing elements one-by-one using loops.

#### Without Vectorization (Python List)

````python


""" Suppose:"""

marks = [10,20,30,40]

""" You want to add 5 to every mark."""

new_marks = []

for i in marks:
    new_marks.append(i + 5)

print(new_marks)
````

### 1.3.3 Why is NumPy faster than Python lists?

1. NumPy stores data in contiguous memory.
2. Arrays contain homogeneous data types.
3. Operations are implemented in optimized C code.
4. NumPy uses vectorized operations instead of Python-level loops.

###  Contiguous Memory

NumPy stores elements in a continuous block of memory.

[10][20][30][40]

This allows faster access and computation compared to Python lists, which store references to objects.

- Faster than zeros()
- Contains garbage values

Why Strange Values?

Because NumPy:

Allocates memory
BUT
Does not fill it

The memory may already contain old data.

```
Shape is: (5,)
It contains 5 memory locations.

A truly empty array is: np.array([])

or

np.empty(0)

Shape: (0,)
```

Slicing is used to extract a portion of an array.
```
Syntax:

array[row_start:row_stop, col_start:col_stop]
```

 1d array

````python
a1
````

````python
print(a1[0:4])
````

````python
print(a1[-4:])
````

2d array

- arr[0:1, :]: Grabs rows starting at 0 up to (but not including) 1.
- arr[0, :]: Grabs exactly row 0.

The general rule for 2D slicing
- sub_array = array[row_start:row_stop:jump, col_start:col_stop:jump]

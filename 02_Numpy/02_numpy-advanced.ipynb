# 3. Broadcasting



Broadcasting is NumPy's mechanism for performing arithmetic operations on arrays of different shapes by automatically expanding the smaller array to match the shape of the larger array.

````python
import numpy as np

arr = np.array([1,2,3])

arr + 5


"""
5 -> [5,5,5]
[1,2,3] + [5,5,5] 

"""
````

## 3.1 Broadcasting Rules

NumPy compares shapes from right to left.

Two dimensions are compatible if:

1. They are equal
OR
2. One of them is 1
```
(3, 2) (3,)
(3, 2)  (1,3)

(3, 3, 3) (3,)
(3, 3, 3) (1, 1, 3)
```

````python
"Example 1"
"""
(3,4)
(3,1)

Compare:

4 can expand from 1 ✔
3 == 3 ✔

Broadcasting works.
"""
````

````python
"Example 2"

"""
(3,4)
(2,4)

Compare:

4 == 4 ✔
3 ≠ 2 ✘

Broadcasting fails.
"""
````

## 3.2 Scalar Broadcasting

````python
a = np.array([1,2,3,4])

print(a + 10)

# Output:

""" [11 12 13 14]"""

# Internally:
"""
[1 2 3 4]
+
[10 10 10 10]
"""
````

## 3.3 2D Array + Scalar

````python
a = np.array([
    [1,2,3],
    [4,5,6]
])

print(a + 100)

# Output:
"""
[[101 102 103]
 [104 105 106]]

Internally:

[
 [1 2 3]
 [4 5 6]
]

+

[
 [100 100 100]
 [100 100 100]
]
"""
````

## 3.4 Row-wise Broadcasting

````python
a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([10,20,30])

print(a + b)

"""
Shapes:

a.shape -> (2,3)
b.shape -> (3,)

NumPy expands:

[10 20 30]

↓

[
 [10 20 30]
 [10 20 30]
]

Then adds.
"""
````

## 3.5 Column-wise Broadcasting

````python
a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([
    [10],
    [20]
])

print(a + b)

"""
Shapes:

a.shape -> (2,3)
b.shape -> (2,1)

NumPy expands:

[
 [10]
 [20]
]

↓

[
 [10 10 10]
 [20 20 20]
]
"""
````

## Broadcasting Fails


````python
a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([
    [10,20],
    [30,40]
])

print(a + b)
````

Vectorization:
- Perform operations on entire arrays without explicit loops.

Broadcasting:
- Allows operations between arrays of different shapes by automatically expanding dimensions.

## 4.2 Means squared error

```
MSE= 1/n ​∑i->n​(yi​−y^​i​)^2

yi = actual value
y^i= predicted value
n = number of samples
```

````python
def mse(actual, predicted):
  return sum((actual - predicted)**2)/len(actual)

actual = np.arange(24,36)
predicted = np.arange(12,24)
# print(actual)
# print(predicted)

mse(actual, predicted)
````

### 3) inverse()

- Not every matrix has an inverse.
```
- Notice:

Row2 = 2 × Row1

Rows are dependent.

Inverse does not exist.
```

A matrix inverse is a matrix that produces the identity matrix when multiplied with the original matrix.

A × A⁻¹ = I

Modern NumPy:

np.linalg.inv(A)

Conditions:

1. Matrix must be square.
2. Determinant must be non-zero.

Related Function:

np.linalg.det()

````python
"Example 1"

a = np.matrix([[3,4],[2,1]])
print(a)
print()
print(np.linalg.inv(a))
````

````python
"Example 2" 

""" Error : rows are dependet."""

a = np.matrix([[1,2],[2,4]])
print(a)
print()
print(np.linalg.inv(a))
````

### 4)  np.linalg.matrix_power()

Raises a square matrix to a power.

Syntax:

np.linalg.matrix_power(A, n)

Cases:

n > 0
→ Matrix multiplied n times

n = 0
→ Identity Matrix

n < 0
→ Inverse matrix raised to power |n|

Important:

Works only for square matrices.

````python
"Example 1"

""" For n = 0"""

a = np.matrix([[1,2],[4,5]])
print(a)
print()
print(np.linalg.matrix_power(a,0))
````

````python
"Example 2"

"""For n > 0"""

a = np.matrix([[1,2],[4,5]])
print(a)
print()
print(np.linalg.matrix_power(a,2))
````

````python
"EExample 3"

"For n < 0"

a = np.matrix([[1,2],[4,5]])
print(a)
print()
print(np.linalg.matrix_power(a,-1))
````

### 5) Determinant

The determinant is a special value calculated from a square matrix.

Notation: det(A) or  |A|

A determinant is a special value associated with a square matrix.

```
Syntax:   np.linalg.det(A)

For 2×2 matrix:

[a b]
[c d]

det = ad - bc
```

Important:

det ≠ 0
→ Inverse exists

det = 0
→ Inverse does not exist
Quick Check


````python
"Example 1"

a = np.matrix([[4,2],[1,2]])

print(a)
print()
print(np.linalg.det(a))
````

````python
"Example 2"

b = np.matrix([[1,2,3],[4,5,6],[7,8,9]])
print(b)
print()
print(np.linalg.det(b))
````

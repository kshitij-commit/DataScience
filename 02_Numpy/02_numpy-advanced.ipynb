# 3. Broadcasting



Broadcasting is NumPy's mechanism for performing arithmetic operations on arrays of different shapes by automatically expanding the smaller array to match the shape of the larger array.

````python
import numpy as np

arr = np.array([1,2,3])

arr + 5


"""
5 -> [5,5,5]
[1,2,3] + [5,5,5] 

"""
````

## 3.1 Broadcasting Rules

NumPy compares shapes from right to left.

Two dimensions are compatible if:

1. They are equal
OR
2. One of them is 1
```
(3, 2) (3,)
(3, 2)  (1,3)

(3, 3, 3) (3,)
(3, 3, 3) (1, 1, 3)
```

````python
"Example 1"
"""
(3,4)
(3,1)

Compare:

4 can expand from 1 ✔
3 == 3 ✔

Broadcasting works.
"""
````

````python
"Example 2"

"""
(3,4)
(2,4)

Compare:

4 == 4 ✔
3 ≠ 2 ✘

Broadcasting fails.
"""
````

## 3.2 Scalar Broadcasting

````python
a = np.array([1,2,3,4])

print(a + 10)

# Output:

""" [11 12 13 14]"""

# Internally:
"""
[1 2 3 4]
+
[10 10 10 10]
"""
````

## 3.3 2D Array + Scalar

````python
a = np.array([
    [1,2,3],
    [4,5,6]
])

print(a + 100)

# Output:
"""
[[101 102 103]
 [104 105 106]]

Internally:

[
 [1 2 3]
 [4 5 6]
]

+

[
 [100 100 100]
 [100 100 100]
]
"""
````

## 3.4 Row-wise Broadcasting

````python
a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([10,20,30])

print(a + b)

"""
Shapes:

a.shape -> (2,3)
b.shape -> (3,)

NumPy expands:

[10 20 30]

↓

[
 [10 20 30]
 [10 20 30]
]

Then adds.
"""
````

## 3.5 Column-wise Broadcasting

````python
a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([
    [10],
    [20]
])

print(a + b)

"""
Shapes:

a.shape -> (2,3)
b.shape -> (2,1)

NumPy expands:

[
 [10]
 [20]
]

↓

[
 [10 10 10]
 [20 20 20]
]
"""
````

## Broadcasting Fails


````python
a = np.array([
    [1,2,3],
    [4,5,6]
])

b = np.array([
    [10,20],
    [30,40]
])

print(a + b)
````

Q. Difference between Vectorization and Broadcasting?

Vectorization:
- Perform operations on entire arrays without explicit loops.

Broadcasting:
- Allows operations between arrays of different shapes by automatically expanding dimensions.

````python
a = np.arange(12).reshape(3,4)
b = np.arange(12,24).reshape(4,3)

print(a)
print("\n" \
"")
print(b)
print(a+b)  # gives error as dimensions are not same. 
````

## 4.1 Sigmoid

```
formula = 1/(1 + e^(-x))
```

````python
def sigmoid(x):
    return 1/(1 + np.exp(-x))


arr = np.arange(12).reshape(3,4)
sigmoid(arr)
````

## 4.2 Means squared error

```
MSE= 1/n ​∑i->n​(yi​−y^​i​)^2

yi = actual value
y^i= predicted value
n = number of samples
```

````python
def mse(actual, predicted):
  return sum((actual - predicted)**2)/len(actual)

actual = np.arange(24,36)
predicted = np.arange(12,24)
# print(actual)
# print(predicted)

mse(actual, predicted)
````

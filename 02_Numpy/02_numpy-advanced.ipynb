# 1. Numpy array vs Python lists

## 1.1 speed

list

```python
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]

c = []
import time

start = time.time()
for i in range(len(a)):
    c.append(a[i] + b[i])

end = time.time() - start
end
```

array

```python
import numpy as np
```

```python
c = np.arange(10000000)
d = np.arange(10000000,20000000)

start = time.time()
d = c+d
end = time.time() - start
end
```

```python
print(2.69/0.04)

"array is 67 times faster than python list"
```

```python
"i want 0th, 2nd, 3rd 5th row"

print(a1[[0,2,3,5]])
```

## 2.2 Boolean Indexing

## random.randint()

```python
a = np.random.randint(1,100,24).reshape(6,4)
a
```

```python
a > 50
```

```python
a[a>50]
```

```python
"Find out even numbers."

a[a % 2 == 0]
```

```python
"Find out numbers greater than 50 and also even. "

a[(a % 2 == 0) & (a>50)]
```

```python
"Find numbers not divisible by 7."

a[a % 7 == 0]
```

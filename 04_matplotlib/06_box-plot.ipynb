# box - whisker plot


A Box Plot is used to visualize the distribution of data, showing:

- Minimum value
- First Quartile (Q1)
- Median (Q2)
- Third Quartile (Q3)
- Maximum value
- Outliers

Minimum  : Smallest value in the dataset.

Q1 (First Quartile) : Median of the lower half (25% of the data lies below it).

Median (Q2) : Middle value of the dataset (50% of the data lies below it).

Q3 (Third Quartile) : Median of the upper half (75% of the data lies below it).

Maximum : Largest value in the dataset.

Outlier : A value that is much smaller or larger than the rest of the data.

## Why do we use a Box Plot?

- Compare data distributions.
- Find the median.
- Detect outliers.
- Understand the spread (variation) of data.

## Syntax
```
plt.boxplot(x)
x → List or array of numerical values.
```

````python
import matplotlib.pyplot as plt

marks = [10,20,30,40,50,60,70]

plt.boxplot(marks)
plt.show()
````

## Explanation

````python
data = [10,20,30,40,50,60,70,80,90]
````

### 1. Minimum Value

The smallest value in the dataset.

- 10

### 2. Median (Q2)

The middle value of the data.

10 20 30 40 [50] 60 70 80 90

So,

Median (Q2) = 50

Think of the median as the value that splits the data into two equal halves.

### 3. First Quartile (Q1)

Now ignore the median.

Left half:

10 20 30 40

Find the median of this half.

20 30

Average:

(20 + 30) / 2 = 25

So,

Q1 = 25

Meaning:

25% of the data lies below Q1.

### 4. Third Quartile (Q3)

Ignore the median again.

Right half:

60 70 80 90

Middle values:

70 80

Average:

(70 + 80) / 2 = 75

So,

Q3 = 75

Meaning:

75% of the data lies below Q3.

### 5. Maximum Value

Largest value.

90

### 6. Outliers ⭐

Now consider:

10 20 30 40 50 60 70 80 500

Everything is between 10 and 80.

Suddenly,

500

appears.

This value is far away from the rest.

That is called an Outlier.

Box Plot shows outliers as separate points:

│──────□──────│                    •
                               500

The dot (•) is an outlier.

## Anatomy of a Box Plot

### 1. Box 📦
      ┌──────────────┐
      │      │       │
      └──────────────┘

The box starts at Q1 and ends at Q3.

So,

Left edge  = Q1
Right edge = Q3

The box contains the middle 50% of the data.

### 2. Median
      ┌────│────┐
      │    │    │
      └────│────┘

The line inside the box is the Median (Q2).

It divides the data into two equal halves.

### 3. Whiskers
```
──────┐
      │
 ┌────┴────┐
 │         │
 └────┬────┘
      │
──────┘
```
These lines are called Whiskers.

- Lower whisker → Extends toward the minimum value.
- Upper whisker → Extends toward the maximum value.

⚠️ Note:
- If outliers exist, whiskers stop at the last non-outlier value, not necessarily the absolute minimum or maximum.

### 4. Outliers
          ●

Dots outside the whiskers are Outliers.

These are values much larger or smaller than the rest of the data.

Example:

10 15 18 20 22 24 25 27 30 120

Here:

120

will likely appear as:

          ●

### 5. Interquartile Range (IQR)



IQR = Q3 - Q1

Example:

Q1 = 20
Q3 = 50

IQR = 50 - 20 = 30
Why is IQR important?

It measures the spread of the middle 50% of the data.

A larger IQR means the middle data is more spread out.

## Topic: notch

A notch adds a V-shaped indentation around the median line.

It helps visually compare the medians of different box plots.

When comparing multiple box plots, notches make it easier to compare their median values.

```
Syntax
plt.boxplot(data, notch=True)
```

````python
import matplotlib.pyplot as plt

marks = [10,20,30,40,50,60,70]

plt.boxplot(marks)

plt.show()
````

````python
import matplotlib.pyplot as plt

marks = [10,20,30,40,50,60,70]

plt.boxplot(marks, notch=True)

plt.show()
````

````python
'''Example '''
class_A = [45,50,55,60,65,70]
class_B = [30,35,40,45,50,55]

plt.boxplot([class_A, class_B], notch=True)

plt.show()


````

## Topic: patch_artist

patch_artist=True allows the box to be filled with color.

Without it, the box is only an outline.

patch_artist=False → Default (only box outline)

patch_artist=True → Allows the box to be filled with color.

Used with:
- boxprops=dict(facecolor='color')

```
Syntax
plt.boxplot(data, patch_artist=True)
```

````python
import matplotlib.pyplot as plt

marks = [10,20,30,40,50,60,70]

plt.boxplot(marks, patch_artist=True)

plt.show()
````

````python
import matplotlib.pyplot as plt

marks = [10,20,30,40,50,60,70]

plt.boxplot(marks, patch_artist=True, boxprops=dict(facecolor= 'r'))

plt.show()
````

## Topic: boxprops

boxprops is used to customize the box of the box plot.

You can change:

- Border color
- Border width
- Border style
- Face color (requires patch_artist=True)

```
Syntax
plt.boxplot(
    data,
    boxprops=dict(...)
)
```

### 1. Change Border Color

````python
marks = [10,20,30,40,50,60,70]

plt.boxplot(marks, boxprops=dict(color='r'))
plt.show()
````

### 2. Change Border Width
```
plt.boxplot(
    marks,
    boxprops=dict(linewidth=3)
)
```
plt.show()

📌 Makes the border thicker.

````python
marks = [10,20,30,40,50]

plt.boxplot(marks, boxprops=dict(linewidth= 3))
plt.show()
````

### 3. Change Border Style
```
plt.boxplot(
    marks,
    boxprops=dict(linestyle='--')
)
```
plt.show()

📌 Dashed border.

````python
marks = [10,20,30,40,50]

plt.boxplot(marks, boxprops=dict(linestyle= '--'))
plt.show()
````

### 4. Fill the Box
```
plt.boxplot(
    marks,
    patch_artist=True,
    boxprops=dict(facecolor='skyblue')
)
```
plt.show()

📌 Fills the box with color.

````python
marks = [10,20,30,40,50]

plt.boxplot(marks,patch_artist=True ,boxprops=dict(facecolor='green'))
plt.show()
````

## labels

````python
marks = [1,2,3,4,5]

plt.boxplot(marks, labels=['apple'])
plt.show()
````

### 5. Combine Everything

````python

plt.boxplot(
    marks,
    patch_artist=True,
    boxprops=dict(
        facecolor='skyblue',
        color='red',
        linewidth=3,
        linestyle='--'
    )
)

plt.show()
````

## showmeans 

Normally, a box plot shows the median only.

If you also want to show the mean (average), use:

howmeans=True  → Show the mean (average)

showmeans=False → Hide the mean (Default)

It does NOT remove the median.

````python
marks = [1,2,3,4,5]

plt.boxplot(marks, labels=['apple'], showmeans=True)
plt.show()
````

## flierprops

flierprops is used to customize the outliers (fliers).

You can change:

- Color
- Marker

```
flierprops → Customize outliers.

marker           → Shape of outlier
markerfacecolor  → Outlier color

```

```
Syntax
```

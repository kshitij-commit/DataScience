# Scatter plot

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [3,4,9,10]

plt.scatter(x,y)

plt.show()
```

## title,  x, y axis labels  and fontsize

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [3,4,9,10]

plt.xlabel("numbers", fontsize = 20)
plt.ylabel('month',  fontsize = 20)
plt.title('scatter plot', fontsize= 20)

plt.scatter(x,y)

plt.show()
```

## color

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [3,4,9,10]

plt.xlabel("numbers", fontsize = 20)
plt.ylabel('month',  fontsize = 20)
plt.title('scatter plot', fontsize= 20)

c = ['r','g','pink','m']
plt.scatter(x,y, color= c)

plt.show()
```

## size (s)

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [3,4,9,10]

plt.xlabel("numbers", fontsize = 20)
plt.ylabel('month',  fontsize = 20)
plt.title('scatter plot', fontsize= 20)

c = ['r','g','pink','m']

size = [200, 500, 800, 150]
plt.scatter(x,y, color= c, s = size)

plt.show()
```

## marker

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [3,4,9,10]

plt.xlabel("numbers", fontsize = 20)
plt.ylabel('month',  fontsize = 20)
plt.title('scatter plot', fontsize= 20)

c = ['r','g','pink','m']

size = [200, 500, 800, 150]
plt.scatter(x,y, color= c, s = size, marker='*', edgecolors='black')

plt.show()
```

## color (cmap) and colorbar

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [3,4,9,10]

plt.xlabel("numbers", fontsize = 20)
plt.ylabel('month',  fontsize = 20)
plt.title('scatter plot', fontsize= 20)



''' For using cmap you have to give numbers to colors'''
c = [10,20,30,40]

size = [200, 500, 800, 150]
plt.scatter(x,y, c = c, s = size, marker='*', edgecolors='black', cmap='Accent')
plt.colorbar()

plt.show()
```

## colorbar label

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [3,4,9,10]

plt.xlabel("numbers", fontsize = 20)
plt.ylabel('month',  fontsize = 20)
plt.title('scatter plot', fontsize= 20)



''' For using cmap you have to give numbers to colors'''
c = [10,20,30,40]

size = [200, 500, 800, 150]
plt.scatter(x,y, c = c, s = size, marker='*', edgecolors='black', cmap='flag')
t = plt.colorbar()
t.set_label('colorbar', fontsize= 20)



plt.show()
```

## multiple scatter plot

```python
import matplotlib.pyplot as plt 

x = [1,2,3,4]
y = [3,4,9,10]
z = [10,20,30,40]

plt.xlabel("numbers", fontsize = 20)
plt.ylabel('month',  fontsize = 20)
plt.title('scatter plot', fontsize= 20)



''' For using cmap you have to give numbers to colors'''
c = [10,20,30,40]

size = [200, 500, 800, 150]
plt.scatter(x,y, c = c, s = size, marker='*', edgecolors='black', cmap='flag')
t = plt.colorbar()
t.set_label('colorbar', fontsize= 20)


plt.scatter(x,z, c = c, s = size, marker='*', edgecolors='black', cmap='binary')
t = plt.colorbar()
t.set_label('colorbar', fontsize= 20)



plt.show()
```

# stem plot

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [1,4,9,16,25]

plt.stem(x, y)
```

## linefmt

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [1,4,9,16,25]

plt.stem(x, y, linefmt=':')
```

## marker (markerfmt)

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [1,4,9,16,25]

plt.stem(x, y, linefmt=':', markerfmt='+')
```

## bottom

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [1,4,9,16,25]

plt.stem(x, y, linefmt=':', markerfmt='o', bottom=2)
```

## basefmt

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [1,4,9,16,25]

plt.stem(x, y, linefmt=':', markerfmt='o', bottom=9, basefmt='g')
```

## legend()

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [1,4,9,16,25]

plt.stem(x, y, linefmt=':', markerfmt='o', bottom=9, basefmt='g', label='code')
# plt.legend(loc='upper right')
plt.legend()
plt.show()
```

## uselinecollection

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [1,4,9,16,25]

plt.stem(x, y, linefmt=':', markerfmt='o', bottom=9, basefmt='g', label='code', use_line_collection=False) 
# plt.legend(loc='upper right')
plt.legend()
plt.show()
```

## orientation

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]

y = [1,4,9,16,25]

plt.stem(x, y, linefmt=':', markerfmt='o', bottom=9, basefmt='g', label='code', orientation='horizontal') 
# plt.legend(loc='upper right')
plt.legend()
plt.show()
```

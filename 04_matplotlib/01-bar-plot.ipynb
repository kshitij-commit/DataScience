# Introduction to Matplotlib

- Matplotlib is a Python library used for data visualization.
- It is used to create graphs, charts, and plots.
- It helps in understanding and analyzing data visually.
- It works well with NumPy and Pandas.
- Most plotting functions are available in the pyplot module.

## Plot Types


1. Linear Plot (plot)
2. Scatter Plot (scatter)
3. Bar Plot (bar)
4. Stem Plot (stem)
5. Step Plot (step)
6. Histogram Plot (hist)
7. Box Plot (boxplot)
8. Pie Plot (pie)
9. Fill Between Plot (fill_between)

### legend

Legend Position:
```
1. plt.legend(loc='upper right')
2. plt.legend(loc='upper left')
3. plt.legend(loc='lower right')
4. plt.legend(loc='lower left')
5. plt.legend(loc='center')
```

A legend is used to identify which color, line, marker, or bar represents which dataset in a graph.

Without legend:

You can see the graph.But you may not know what each color or line means.

With legend:

The graph explains itself.

### multigraph


````python
x = ["car",'bike','scootey','bicycle']
y = [20,30,35,15]
z = [10, 25, 20, 10]


plt.xlabel('Type of Vehicles')
plt.ylabel('Number of vehicles')

plt.legend(loc= 'upper right')
plt.bar(x,y, color='r',width=0.4,edgecolor='black', label='vehicle y')
plt.bar(x,z,color='g',width=0.4,edgecolor='black', label= 'vehicle z')

plt.legend(loc='upper right')

plt.title('Praking Survey')
plt.show()

````

### separate bars 

Separate karne ke liye:

- Ya to pehli bar ko thoda left shift karo.
- Ya dusri bar ko thoda right shift karo.

Original Position:    x

After Separation:
- First Bar  -> x - something
- Second Bar -> x + something

```
When x-axis contains string labels, create numeric positions using np.arange().
Shift the numeric positions and use plt.xticks() to display original labels.
```

```
General Formula (Very Important ⭐)

If you write:

plt.bar(position, height, width=w)

Then:

Left Edge = position
Right Edge = position + w
Center = position + (w / 2)
```

````python
import numpy as np
````

````python
x = ["car",'bike','scootey','bicycle'] 
''' We can't shift x position because it is string'''
y = [20,30,35,15]
z = [10, 25, 20, 10]

x_pos = np.arange(len(x))  
# x_pos


plt.xlabel('Type of Vehicles')
plt.ylabel('Number of vehicles')

plt.legend(loc= 'upper right')
plt.bar(x_pos-0.4,y, color='r',width=0.4,edgecolor='black', label='vehicle y')
plt.bar(x_pos,z,color='g',width=0.4,edgecolor='black', label= 'vehicle z')



plt.legend(loc='upper right')

plt.title('Praking Survey')
plt.show()



````

### xticks()

Use plt.xticks() to display the original string labels.

````python
x = ["car",'bike','scootey','bicycle'] 
''' We can't shift x position because it is string'''
y = [20,30,35,15]
z = [10, 25, 20, 10]

x_pos = np.arange(len(x))  
# x_pos

plt.xticks(x_pos,x)

plt.xlabel('Type of Vehicles')
plt.ylabel('Number of vehicles')


plt.bar(x_pos-0.4,y, color='r',width=0.4,edgecolor='black', label='vehicle y')
plt.bar(x_pos,z,color='g',width=0.4,edgecolor='black', label= 'vehicle z')

plt.legend(loc= 'upper right')




plt.title('Parking Survey')
plt.show()



````

General Rule:

- Left Bar Position  = P1
- Right Bar Position = P2

- Tick Position = (P1 + P2)/2

````python
x = ["car",'bike','scootey','bicycle'] 
''' We can't shift x position because it is string'''
y = [20,30,35,15]
z = [10, 25, 20, 10]

x_pos = np.arange(len(x))  
# x_pos

plt.xticks(x_pos-0.2,x)

plt.xlabel('Type of Vehicles')
plt.ylabel('Number of vehicles')


plt.bar(x_pos-0.4,y, color='r',width=0.4,edgecolor='black', label='vehicle y')
plt.bar(x_pos,z,color='g',width=0.4,edgecolor='black', label= 'vehicle z')

plt.legend(loc= 'upper right')




plt.title('Parking Survey')
plt.show()



````

### rotation

````python
import numpy as np
import matplotlib.pyplot as plt 
````

````python
x = ["car",'bike','scootey','bicycle'] 
''' We can't shift x position because it is string'''
y = [20,30,35,15]
z = [10, 25, 20, 10]

x_pos = np.arange(len(x))  
# x_pos

plt.xticks(x_pos-0.2,x, rotation= 30)

plt.xlabel('Type of Vehicles')
plt.ylabel('Number of vehicles')


plt.bar(x_pos-0.4,y, color='r',width=0.4,edgecolor='black', label='vehicle y')
plt.bar(x_pos,z,color='g',width=0.4,edgecolor='black', label= 'vehicle z')

plt.legend(loc= 'upper right')




plt.title('Parking Survey')
plt.show()



````

### plt.barh() (Horizontal Bar Plot)

plt.barh() is used to create horizontal bar charts.

````python
import matplotlib.pyplot as plt

subjects = ['Math', 'Physics', 'Chemistry', 'English']
marks = [85, 72, 90, 78]

plt.barh(subjects, marks)

plt.xlabel("Marks")
plt.ylabel("Subjects")
plt.title("Student Marks")

plt.show()
````

````python
x = ['Jan', 'Feb', 'Mar', 'Apr']
sales_2024 = [20, 30, 25, 35]
sales_2025 = [25, 35, 30, 40]

x_pos = np.arange(len(x))

plt.bar(x_pos-0.4, sales_2024, color = 'r', width=0.4)
plt.bar(x_pos, sales_2025, color = 'g',width=0.4)

plt.xticks(x_pos-0.2, x)
plt.show()
````

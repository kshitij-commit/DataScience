# Histogram

````python
import matplotlib.pyplot as plt
import numpy as np
````

````python
x = np.random.randint(10,60,(50))
````

````python
plt.hist(x)
plt.show()
````

## labels and colors

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r')
plt.show()
````

## edgecolor

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black')
plt.show()
````

## bins

````python
import matplotlib.pyplot as plt
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")

l = [10,20,30,40,50,60,70,80]

plt.hist(x, color='r',edgecolor='black', bins=l)
plt.show()
````

## increase x axis range

bins='auto' → Number of bins automatically choose karta hai.

range=(0,100) → Histogram kis interval me banega, ye decide karta hai.

```
bins='auto' :
Matplotlib automatically decides the number of bins based on the data.

range=(min, max) :
Specifies the interval used to create the histogram.
Only values within this range are considered; values outside the range are ignored.
```

```
plt.hist(data, bins='auto', range=(0,100))
bins='auto' → Automatically chooses the number of bins.
range=(0,100) → Only data between 0 and 100 is used to create the histogram.
```

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")

# l = [10,20,30,40]

plt.hist(x, bins='auto',range=(0,100), color='r',edgecolor='black')

# plt.xlim(0, 200)

plt.show()
````

## cumulative

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', cumulative=-1)
plt.show()
````

## bottom

Sets the starting values of x and y axis.

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', bottom=10)
plt.show()
````

## histtype = step


````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', histtype='step')
plt.show()
````

## histtype = stepfilled

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', histtype='stepfilled')
plt.show()
````

## histtype = barstacked

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', histtype='barstacked')
plt.show()
````

## Orientation

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', histtype='barstacked', orientation='horizontal')
plt.show()
````

## rwidth

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', histtype='barstacked', rwidth=0.6)
plt.show()
````

## log

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', histtype='barstacked', log=True)
plt.show()
````

## legend

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', label='code')

plt.legend()
plt.show()
````

## axvline

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', label='code')

plt.legend()

plt.axvline(45, color= 'black')

plt.show()
````

## axvline (label)

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', label='code')

plt.axvline(45, color= 'black', label='separate')
plt.legend()


plt.show()
````

## grid function

````python
plt.title("Record")
plt.xlabel("Python")
plt.ylabel("Number")


plt.hist(x, color='r',edgecolor='black', label='code')

plt.axvline(45, color= 'black', label='separate')

plt.grid()
plt.legend()


plt.show()
````
